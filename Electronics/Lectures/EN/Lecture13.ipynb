# **Electronic Amplifiers: Operational Amplifiers (op-amp)**

## **Electronic Amplifiers**

<img src="../figs/lec9/960px-McIntosh_MC2505.jpg" alt="Vdiode" width="50%" style="background-color:#fff;" >


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">
    
#### 1. What is an amplifier?

An **electronic amplifier** (or simply an “amp”) is a two-port circuit that uses power from a supply to produce an output signal with a larger amplitude than the input signal. The amount of increase is called **gain** – the ratio of output to input voltage, current, or power. In the full physical sense, an amplifier must have a **power gain greater than 1**, because the additional output energy comes from the external power supply.

</div>


#### 2. Main classifications

| Based on                 | Example / Range                                            | Main use                                                       |
| ------------------------ | ---------------------------------------------------------- | -------------------------------------------------------------- |
| **Frequency**            | Audio (< 20 kHz), RF (20 kHz – 300 GHz), DC/servo          | Hi-fi devices, radio transmitters, motor drives |
| **Position in chain**    | *Preamplifier* (before processing), *power amp* (at the end) | Sensor conditioning vs. loudspeaker driving |
| **Conduction class**     | A, B, AB, C (linear); D, E (switching)                     | Compromise between efficiency and linearity |
| **Phase configuration**  | Inverting, non-inverting                                   | Sets the input-output phase relation       |


#### 3. Active devices that provide amplification

* **Vacuum tubes (triodes, pentodes)** – the origin of amplifiers (1906-1960s).
* **BJT & MOSFET transistors** – today they make up most amplifiers.
* Special RF devices (klystron, TWT) or **negative-resistance** devices (tunnel diode) for specific applications.


#### 4. Main performance parameters

| Parameter                          | What it measures                                  | Why it matters                           |
| ---------------------------------- | ------------------------------------------------- | ---------------------------------------- |
| **Gain ($A_v$, $A_i$, $A_p$)**     | Output/input ratio                                | How much the signal “increases”          |
| **Bandwidth (BW)**                 | Frequency where the gain drops by 3 dB            | Signal speed                             |
| **Efficiency**                     | Output power / total consumption                  | Critical for power amplifiers            |
| **Linearity & Distortion**         | How accurately the waveform is copied             | Hi-fi audio requires THD < 0.01 %        |
| **Noise**                          | Input voltage/current noise                       | Limit for very small signals             |
| **Slew Rate, stability**           | Speed limitation & avoidance of self-oscillation  | Sets practical limits                    |



#### 5. Operating principle

The amplifier is represented as a **dependent voltage or current source**. In practice, most amplifiers use **negative feedback**: part of the output is returned to the input with opposite phase in order to reduce the open-loop gain, but to **increase bandwidth and reduce distortion**.


#### 6. Common types

* **Audio power amp** – raises a weak audio signal to watts in order to drive loudspeakers.
* **RF power amp** – the final stage of a transmitter, optimized for power transfer to the antenna.
* **Preamplifier/instrumentation amplifier** – amplification of μV–mV voltages with ultra-low noise.
* **Operational amplifier (op-amp)** – a differential “amplifier block” with very large open-loop gain; its characteristics are set by the external feedback circuit.


#### 7. Typical applications

Telephones, hi-fi audio, guitar amplifiers, radar systems, measuring instruments, analog computers, servomotors: all depend on some form of amplifier to process or generate signals at usable levels.


#### 8. Key points to remember

1. **Amplification** does not create energy – the additional energy comes from the external supply.
2. Performance is determined by **gain, bandwidth, linearity, noise, efficiency**.
3. **Conduction class** and **negative feedback** are the main tools for balancing efficiency and signal quality.
4. **Transistors** have replaced the vacuum tube in most applications, but the fundamental techniques remain the same.

## **Operational amplifier (op-amp)**

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

### **What is an op-amp?**

An **op-amp** (operational amplifier) is an electronic component (integrated circuit) used to **amplify the voltage difference** between two points. In most linear applications, it is used as part of an **external feedback circuit**, which determines its function.

> It is called "operational" because it was first used in **analog computers** to perform **mathematical operations such as addition, subtraction, integration, differentiation, etc.**

</div>

<img src="../figs/lec9/LM741CN.jpg" width="30%" style="background-color:#fff;" >


### **How does an op-amp operate?**

**It has two inputs and one output**

<img src="../figs/lec9/Generic_741_pinout_top.png" width="50%" style="background-color:#fff;" >

* **$V^+$** — the **non-inverting** input
* **$V^-$** — the **inverting** input
* **$V_\text{out}$** — the output (usually a voltage)
* **$V_{s+}$** — the positive supply (often not shown in schematics)
* **$V_{s-}$** — the negative supply (often not shown in schematics)

Symbol:
<img src="../figs/lec9/Op-amp_symbol.svg" width="30%" style="background-color:#fff;" >

<br>

The output produces a voltage equal to:
$$
V_\text{out} = A_0\, (V^+ - V^-)
$$

where $A_0$ is the **open-loop gain** – usually **very large** (for example, of order $10^5$ or more).


#### **Operation of the open-loop op-amp**

If we operate an op-amp without feedback, its function is to amplify the voltage difference $V^+ - V^-$. Since $A_0$ is very large (ideally infinite), even a very small difference between $V^+$ and $V^-$ drives $V_\text{out}$ toward very large positive or negative values. But the op-amp is connected to the external supplies $V_{s+}$ and $V_{s-}$; therefore the output voltage cannot exceed the supply limits and goes into saturation. In practice, the output often does not reach the supply rails exactly, but saturates close to them.
Therefore:
* If $V^+ - V^- > 0 \Rightarrow V_\text{out} \approx V_{s+}$
* If $V^+ - V^- < 0 \Rightarrow V_\text{out} \approx V_{s-}$

To create useful applications with op-amps, we use **feedback**, more specifically **negative feedback**.

<br>


## **Feedback**?

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">
    
In a system, **feedback** is the process of returning part of the output signal back to the input.

This “backward” connection helps the system **self-regulate**, **reinforce itself**, or **stabilize itself** — depending on how it is implemented.

</div>

> In an electronic circuit with an op-amp, feedback is the way we make the op-amp behave as an amplifier, integrator, filter, etc.


### Types of feedback

1. **Negative feedback**
    > The output is fed back to the input with **opposite phase**.
    * It aims to **reduce the difference** between the input signal and the returned signal.
    * Net effect: **stability**, **linearity**, **gain control**, **reduction of distortion and noise**.
    * **Used in:**
      * Linear amplifiers
      * Voltage regulators
      * Active filters
      * Control systems (servo, audio), etc.

2. **Positive feedback**
    > The output is fed back to the input with the **same phase**.
    * It **increases the difference** between the input and the feedback signal.
    * This drives the system toward **self-reinforcement** and **saturation** → causing turn-on or rapid transition to another state.
    * **Main effect:** it makes the system **nonlinear** and causes it to **self-activate** once a threshold is crossed.
    * **Used in:**
      * Comparators with hysteresis (Schmitt trigger)
      * Multivibrators (square-wave generators)
      * Oscillators (RC, LC, Wien Bridge), etc.

### Why is feedback important?

| Effect              | Negative feedback           | Positive feedback       |
| ------------------- | --------------------------- | ----------------------- |
| **Stability**       | Improves                    | Can destabilize         |
| **Linearity**       | Improves                    | Decreases               |
| **Noise**           | Decreases                   | Can increase            |
| **Self-activation** | No                          | Yes (circuit turns itself on) |
| **Typical use**     | Amplifiers, filters, servo  | Oscillators, comparators | 


**Feedback** is one of the most fundamental ideas in electronics:
* **Negative feedback** is **the heart of controlled electronics** — it allows us to build stable and accurate systems.
* **Positive feedback** is **the engine of triggering** — from logic transitions to self-generated signals (oscillators).

**In op-amp circuits, every function is connected to how we use feedback.** Therefore, understanding and controlling feedback is the key to understanding and designing electronic circuits.

## **Rules for analyzing circuits with an ideal op-amp**

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">


1. **Rule 1: Open-loop gain of the ideal op-amp $A_0 \to \infty$**
   For an **ideal op-amp**, the open-loop voltage gain is **infinite**, $ A_0 = \infty $. For a **real op-amp**, the gain is large but **finite**, usually in the range: $ 10^4 \text{ to } 10^6 $

2. **Rule 2: Input impedance ($ R_\text{in} \to \infty $) and output impedance ($ R_\text{out} \to 0 $)**
    For an **ideal op-amp**, the **input impedance** is infinite, $ R_\text{in} = \infty $, while the output impedance is zero, $ R_\text{out} = 0 $. For a **real op-amp**, the input impedance is high but **finite**, usually from $10^6\, \Omega$ (e.g. a bipolar-input op-amp) to $10^{12} \, \Omega$ (e.g. a JFET-input op-amp). The output impedance is usually in the range $ R_\text{out} \approx 10 \text{ to } 1000\, \Omega $.

3. **Rule 3: Input current at the op-amp terminals is zero**
   In an **ideal op-amp**, the input terminals **draw no current**, $ I^+ = I^- = 0 $. In practice, this is a **very good approximation** because input currents are usually very small: **~pA** (e.g. in JFET-input op-amps) and **~nA** (e.g. in bipolar-input op-amps). Therefore, in most analyses these currents are **ignored without introducing a large error**.

4. **Rule 4: Behavior of the op-amp with negative feedback ($ V^+ \approx V^-$)**
    If the op-amp operates in its linear region and has negative feedback, it **senses a voltage difference** between the non-inverting input $V^+$ and the inverting input $V^-$, and **responds** by changing the output in the direction that **reduces this difference**. Because the gain is very large, at equilibrium we have $ V^+ \approx V^- $. This is often called a **virtual short**, but it does not mean that the two inputs are physically connected.

</div>

These rules are the foundation for analyzing all op-amp circuits in the linear regime — from amplifiers to filters and integrators, etc.

### **How does feedback work in practice?**

#### Example 1: **Non-inverting amplifier**

<img src="../figs/lec9/Op-Amp_Non-Inverting_Amplifier.svg" width="40%" style="background-color:#fff;" >

In this configuration, the input is applied to $V^+ =V_{in}$, and part of the output is returned to $V^-$ through a voltage divider.
$$ V^- = \frac{R_1}{R_1+R_2}\,V_{out}= \beta\,V_{out} $$

From the op-amp gain formula we have
$$
V_{out} = A_0\, (V^+ - V^-) = A_0\, (V_{in} - \beta\,V_{out}) = \left( \frac{1}{\beta + \frac{1}{A_0}} \right)\, V_{in}
$$

Since $A_0$ is very large, then $1/A_0\approx 0$; therefore

$$
V_{out} \approx \frac{V_{in}}{\beta} = \left( 1 + \frac{R_2}{R_1} \right)\, V_{in}
$$

**Gain coefficient:**
$$
A_\text{closed-loop} = 1 + \frac{R_2}{R_1}
$$

Benefit: We can build an amplifier with **stable gain**, determined only by the **resistors** and not by the detailed characteristics of the op-amp.



### Example 2: **Inverting amplifier**

<img src="../figs/lec9/Op-Amp_Inverting_Amplifier.svg" width="40%" style="background-color:#fff;" >

The input is applied to $V^-$, while $V^+=0$ is connected to ground as shown in the figure.

$$ V^- = \frac{1}{R_f+R_{in}}\,(R_f\,V_{in} + R_{in}\,V_{out})$$

From the op-amp gain formula we have
$$
V_{out} = A_0\, (V^+ - V^-) = -A_0\, V^- = -V_{in}\,\frac{A_0\,R_f}{R_f+R_{in}(1+A_0)}
$$

Again, since $A_0\to \infty$ is very large, we obtain
$$
V_{out} \approx -V_{in}\,\frac{R_f}{R_{in}}
$$

**Gain coefficient:**
$$
A_\text{closed-loop} = - \frac{R_f}{R_{in}}
$$

Benefit: It gives an **inverted phase**, precise gain control, and can be used for **summing voltages**.



### **What is an op-amp useful for?**

| Function                        | Description                                          | Use                                      |
| ------------------------------- | ---------------------------------------------------- | ---------------------------------------- |
| **Amplifier**                   | Increases very weak signals                         | Sensors, microphones                     |
| **Buffer (follower)**           | Preserves voltage with low output impedance          | Interfaces for delicate signals          |
| **Active filters**              | Filters signals according to frequency              | Audio, sensors                           |
| **Comparator**                  | Decides whether one signal is larger than another   | Alarm systems, zero-crossing             |
| **Integrator / Differentiator** | Analog mathematical functions                       | Automatic control, signal processing     |
| **Voltage summer**              | Adds several input signals into one output          | Audio mixing, instruments                |
| **Oscillators**                 | Signal generators                                   | Signal generators, e.g. electronic music |


### **Why is it used so much?**

* It is **cheap** and widely available
* It is **predictable** and **easy to model**
* It can perform many different functions with simple circuit changes
* It takes **little space** and can be integrated into almost any electronic system


### **Summary**
In a linear circuit with negative feedback, the op-amp changes its output in the direction that reduces the difference $V^+ - V^-$. Because the gain is very large, this difference becomes very small, and in ideal analysis we take $V^+ \approx V^-$.
*The op-amp does not amplify the input signal by itself — it uses energy from the power supply to build a new signal that imitates the input with controlled gain.*
