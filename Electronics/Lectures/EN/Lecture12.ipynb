## Main idea of Lecture 12

In Lectures 10 and 11 we studied how transistors work and how they are modeled.

**How do we place a transistor at a suitable operating point, and how do we verify the circuit using calculations, simulation, and measurements?**

# **I — The Quiescent Point (Q-Point) and Biasing**

## **1. What is the Q-point?**

The **quiescent point** is the DC state of the transistor when no AC input signal is applied.

For a BJT, the Q-point is usually described by:

$$
Q_{BJT} = (I_C, V_{CE})
$$

For a MOSFET, the Q-point is usually described by:

$$
Q_{MOSFET} = (I_D, V_{DS})
$$

The Q-point determines the operating region of the transistor and how much signal swing is available before **clipping**, deep saturation, or cutoff occurs.

## **2. Why is biasing needed?**

The input signal is usually AC and changes with time. But the transistor is a nonlinear device.  
Therefore, we first establish a stable DC state:

$$
\text{real circuit} = \text{DC biasing} + \text{small AC signal}
$$

Without proper biasing, the transistor may be:

* in **cutoff** and not conducting;
* in **deep saturation** as a switch;
* at an operating point that is unstable with temperature or component tolerances;
* in a region where the signal becomes distorted.

## **3. Where should the Q-point be placed?**

For a **BJT amplifier**, the Q-point is usually placed in the **forward-active region**.

For a **MOSFET amplifier**, the Q-point is usually placed in the **MOSFET saturation region**.

For a **switch**, the Q-point is not placed in the middle of the linear amplification region. The switch moves between:

$$
\text{OFF} \leftrightarrow \text{ON}
$$

that is, between cutoff and deep saturation for a BJT, or between cutoff and triode operation with small $R_{DS(on)}$ for a MOSFET.

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

## Terminology warning: “saturation” in BJTs and MOSFETs

The term **saturation** does not have the same physical meaning in BJTs and MOSFETs.

* In a **BJT**, saturation means that both junctions are forward-biased; the transistor is usually used as a **closed switch**.
* In a **MOSFET**, saturation is the region where the channel is pinched near the drain and the current is mainly controlled by $V_{GS}$; this is often the **amplification** region.

Therefore, when analyzing circuits, always specify the device and the physical operating region.

</div>


# **II — BJT Biasing**

## **1. Voltage-divider biasing**

This is the most common technique for simple BJT amplifiers.

The circuit contains:

* $R_1$ and $R_2$, which set the base voltage $V_B$;
* $R_E$, which stabilizes the emitter current through negative feedback;
* $R_C$, which converts collector-current variations into an output voltage.

As a first approximation:

$$
V_B \approx V_{CC}\frac{R_2}{R_1 + R_2}
$$

$$
V_E \approx V_B - V_{BE}
$$

$$
I_E \approx \frac{V_E}{R_E}
$$

and, for sufficiently large $\beta$:

$$
I_C \approx I_E
$$

The collector-emitter voltage is:

$$
V_{CE} = V_C - V_E
$$

where

$$
V_C = V_{CC} - I_C R_C
$$

therefore:

$$
V_{CE} = V_{CC} - I_C R_C - I_E R_E
$$

If $I_C \approx I_E$:

$$
V_{CE} \approx V_{CC} - I_C(R_C + R_E)
$$

## **2. Checking the operating region**

After the Q-point is calculated, we must check whether the BJT is in the forward-active region.

For an NPN amplifier:

$$
V_{BE} \approx 0.6-0.7\,\text{V}
$$

and we should have:

$$
V_C > V_B > V_E
$$

or, in practical terms:

$$
V_{CE} > V_{CE(sat)}
$$

usually with enough margin, not just slightly above $0.2\,\text{V}$.

## **3. Why is $R_E$ used?**

The emitter resistor creates negative feedback.

If temperature rises, $I_C$ tends to increase. But an increase in $I_E$ also increases $V_E = I_E R_E$, which effectively reduces $V_{BE}$. This opposes the current increase.

Thus $R_E$ makes the Q-point more stable against:

* temperature changes;
* variations in $\beta$;
* differences between real transistor samples.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **BJT summary**

For a BJT amplifier, good biasing should not strongly depend on $\beta$, because $\beta$ varies significantly from one transistor to another.  
The most stable simple technique is usually **base voltage-divider biasing plus an emitter resistor**.

</div>


## **4. Example: Q-point design for a BJT**

Design an operating point for a common-emitter amplifier.

Given:

* $V_{CC} = 12\,\text{V}$
* $R_1 = 100\,\text{k}\Omega$
* $R_2 = 18\,\text{k}\Omega$
* $R_C = 4.7\,\text{k}\Omega$
* $R_E = 1\,\text{k}\Omega$
* $V_{BE} \approx 0.7\,\text{V}$

### **Step 1 — Base voltage**

$$
V_B = 12\,\text{V}\cdot \frac{18}{100+18}
$$

$$
V_B \approx 1.83\,\text{V}
$$

### **Step 2 — Emitter voltage**

$$
V_E = V_B - V_{BE}
$$

$$
V_E \approx 1.83 - 0.7 = 1.13\,\text{V}
$$

### **Step 3 — Emitter and collector current**

$$
I_E = \frac{V_E}{R_E}
$$

$$
I_E \approx \frac{1.13\,\text{V}}{1\,\text{k}\Omega} = 1.13\,\text{mA}
$$

$$
I_C \approx I_E \approx 1.13\,\text{mA}
$$

### **Step 4 — Collector voltage**

$$
V_C = V_{CC} - I_C R_C
$$

$$
V_C \approx 12 - (1.13\,\text{mA})(4.7\,\text{k}\Omega)
$$

$$
V_C \approx 6.69\,\text{V}
$$

### **Step 5 — Collector-emitter voltage**

$$
V_{CE} = V_C - V_E
$$

$$
V_{CE} \approx 6.69 - 1.13 = 5.56\,\text{V}
$$

This is a reasonable Q-point for amplification, because the transistor is far from cutoff and far from deep saturation.


# **III — MOSFET Biasing**

## **1. Biasing with a source resistor**

For an n-channel MOSFET, if the gate is set at a voltage $V_G$ and the source has a resistor $R_S$, then:

$$
V_S = I_D R_S
$$

and

$$
V_{GS} = V_G - V_S = V_G - I_D R_S
$$

In the MOSFET saturation region, the simple model gives:

$$
I_D = \frac{1}{2}k(V_{GS} - V_{th})^2
$$

Combining these two relations gives a nonlinear equation:

$$
I_D = \frac{1}{2}k(V_G - I_D R_S - V_{th})^2
$$

Therefore, for MOSFETs, the Q-point is often found using:

* numerical solution;
* load-line analysis;
* SPICE;
* iteration with a calculator or Python.

## **2. Checking the operating region**

After $I_D$ is found, calculate:

$$
V_S = I_D R_S
$$

$$
V_D = V_{DD} - I_D R_D
$$

$$
V_{DS} = V_D - V_S
$$

For an n-channel MOSFET in saturation:

$$
V_{GS} > V_{th}
$$

and

$$
V_{DS} \geq V_{GS} - V_{th}
$$

If this condition is not satisfied, the MOSFET is in the triode/ohmic region, not in saturation.

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

## Practical warning about $V_{th}$

$V_{th}$ is not the voltage at which the MOSFET is “fully on”.  
In datasheets, $V_{th}$ is usually measured at a very small current, for example $I_D = 250\,\mu\text{A}$.

For switching, you must check **$R_{DS(on)}$** at the real gate-drive voltage, such as $V_{GS}=3.3\,\text{V}$, $5\,\text{V}$, or $10\,\text{V}$.

</div>


# **IV — Practical Procedure for Analyzing Transistor Circuits**

<div style="border:3px solid #268; background-color:#eef8ff; color:#023; border-radius: 12px; padding: 12px;">

## Practical procedure for analyzing transistor circuits

1. Remove the AC signal and analyze the DC circuit.
2. Find the quiescent point:
   * $I_C$, $V_{CE}$ for a BJT;
   * $I_D$, $V_{DS}$ for a FET/MOSFET.
3. Check the operating region.
4. Replace the transistor with the appropriate small-signal model.
5. Calculate gain, input resistance, and output resistance.
6. Check signal-swing limits, clipping, and power dissipation.
7. Compare with SPICE simulation and/or laboratory measurements.

</div>

## **1. DC/AC separation**

In practical analysis, we often use the decomposition:

$$
v(t) = V_{DC} + v_{ac}(t)
$$

For example:

$$
v_{BE}(t) = V_{BE,Q} + v_{be}(t)
$$

and

$$
i_C(t) = I_{C,Q} + i_c(t)
$$

The Q-point determines the parameters of the small-signal model:

For a BJT:

$$
g_m = \frac{I_C}{V_T}
$$

$$
r_\pi = \frac{\beta}{g_m}
$$

For a MOSFET in saturation:

$$
g_m \approx \frac{2I_D}{V_{GS}-V_{th}}
$$

or

$$
g_m \approx \sqrt{2kI_D}
$$

These quantities are not universal constants; they depend on the Q-point.


# **V — Practical Applications**

## **1. BJT as an LED switch**

**Problem:** An LED requires $10\,\text{mA}$ to light. Its forward voltage is approximately $2\,\text{V}$. Use an NPN BJT to turn on the LED when an input signal $V_{in}=5\,\text{V}$ is applied.

Given:

* $V_{CC} = 5\,\text{V}$
* $I_C = 10\,\text{mA}$
* $V_{LED} \approx 2\,\text{V}$
* $V_{CE(sat)} \approx 0.2\,\text{V}$
* emitter connected to ground;
* LED and resistor in the collector branch.

### **LED resistor**

$$
V_R = V_{CC} - V_{LED} - V_{CE(sat)}
$$

$$
V_R = 5 - 2 - 0.2 = 2.8\,\text{V}
$$

$$
R = \frac{V_R}{I_C}
$$

$$
R = \frac{2.8\,\text{V}}{10\,\text{mA}} = 280\,\Omega
$$

In practice, a standard value may be chosen:

$$
R = 270\,\Omega \quad \text{or} \quad 330\,\Omega
$$

### **Base resistor**

For switching, we usually do not use the nominal $\beta$, because we want the BJT to enter saturation reliably.  
Instead, we use a smaller **forced beta**, for example:

$$
\beta_{forced} = 10
$$

Therefore:

$$
I_B = \frac{I_C}{\beta_{forced}}
$$

$$
I_B = \frac{10\,\text{mA}}{10} = 1\,\text{mA}
$$

$$
R_B = \frac{V_{in} - V_{BE}}{I_B}
$$

$$
R_B = \frac{5 - 0.7}{1\,\text{mA}} = 4.3\,\text{k}\Omega
$$

In practice, choose:

$$
R_B \approx 4.7\,\text{k}\Omega
$$

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

#### Warning

If $\beta=100$ is used for the base-current calculation, one obtains $R_B \approx 43\,\text{k}\Omega$.  
This may not drive the transistor safely into saturation, because $\beta$ varies strongly and decreases under real switching conditions.  
For BJT switches, a **forced beta** is usually used.

</div>


## **2. MOSFET as a switch**

For MOSFET switching, the main criterion is not $V_{th}$, but:

$$
R_{DS(on)}
$$

at the real gate-drive voltage.

If a MOSFET is driven by a microcontroller at $3.3\,\text{V}$, choose a MOSFET whose $R_{DS(on)}$ is specified for:

$$
V_{GS}=2.5\,\text{V}
$$

or

$$
V_{GS}=3.3\,\text{V}
$$

If the datasheet gives $R_{DS(on)}$ only for $V_{GS}=10\,\text{V}$, that MOSFET may not be suitable for direct microcontroller drive.

The power dissipated in the MOSFET in the ON state is:

$$
P_{MOSFET} = I_D^2 R_{DS(on)}
$$

This must be compared with the thermal limits of the package.


# **VI — Advanced Models and SPICE Simulation**

## **1. Why are advanced models needed?**

Simple models such as Ebers–Moll, hybrid-$\pi$, or the square-law MOSFET model are very useful for understanding circuits.

However, for real simulations they are not always sufficient, because real transistors have:

* temperature dependence;
* parasitic capacitances;
* high-frequency effects;
* short-channel effects;
* nonlinearities at high currents;
* manufacturing tolerances.

## **2. Advanced BJT models**

| Model | Main use |
|---|---|
| Gummel–Poon | Practical SPICE model for BJTs with better accuracy than Ebers–Moll |
| VBIC | More advanced model for BJTs in integrated circuits |
| HICUM | Model for high currents and very high frequencies |

## **3. Advanced MOSFET models**

| Model | Main use |
|---|---|
| BSIM3/BSIM4/BSIM6 | Standard models for CMOS |
| EKV | Very useful for analog design and subthreshold operation |
| HiSIM | Physics-based model for modern technologies |

## **4. SPICE syntax examples**

### BJT

```spice
Q1 C B E QNPN
.model QNPN NPN (IS=1e-15 BF=100 VAF=100)
```

### MOSFET

```spice
M1 D G S 0 NMOS
.model NMOS NMOS (VTO=1 KP=50u LAMBDA=0.01)
```

## **5. Main SPICE analyses**

| Analysis | Purpose |
|---|---|
| `.op` | Finds the DC operating point |
| `.dc` | Produces static $I$–$V$ characteristics |
| `.ac` | Analyzes small-signal gain and phase |
| `.tran` | Analyzes time-domain response |
| `.step` | Sweeps parameters to study sensitivity |

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary**

Advanced models do not replace simple conceptual models.  
They are used when we need higher numerical accuracy, agreement with real components, and reliable simulation under practical conditions.

</div>


# **VII — Practical Use and Laboratory Exercises with Transistors**

## **1. Measuring BJT $I$–$V$ characteristics**

### BJT — $I_C$ versus $V_{CE}$ characteristics for different $I_B$ values

#### Equipment

* NPN BJT, for example 2N3904 or BC547
* Base current-limiting resistor
* Collector load resistor
* DC power supply
* Multimeter or current sensor
* Oscilloscope, if dynamic observation is required

#### Procedure

1. Fix $I_B$ at a small value, for example $10\,\mu\text{A}$.
2. Gradually vary $V_{CE}$.
3. Measure $I_C$ for each value of $V_{CE}$.
4. Repeat for several values of $I_B$.
5. Plot the family of curves $I_C(V_{CE})$.

#### Result

The curves show:

* cutoff;
* saturation;
* forward-active operation;
* the Early effect if the measurements are sufficiently precise.

## **2. Measuring MOSFET $I$–$V$ characteristics**

### MOSFET — $I_D$ versus $V_{DS}$ characteristics for different $V_{GS}$ values

#### Equipment

* n-channel MOSFET
* DC source for $V_{GS}$
* DC source for $V_{DS}$
* Current-limiting resistor or load
* Multimeter/ammeter

#### Procedure

1. Fix $V_{GS}$ above $V_{th}$, for example $2.5\,\text{V}$, $3.3\,\text{V}$, or $4\,\text{V}$.
2. Gradually vary $V_{DS}$.
3. Measure $I_D$.
4. Repeat for each value of $V_{GS}$.
5. Plot the family of curves $I_D(V_{DS})$.

#### Result

The curves show:

* the triode/ohmic region;
* the MOSFET saturation region;
* the dependence of current on $V_{GS}$;
* channel-length modulation if a slope is visible in the saturation region.


## **3. Testing amplifiers**

### Common-emitter amplifier

1. Build the circuit with a base voltage divider and an emitter resistor.
2. Measure the Q-point: $V_B$, $V_E$, $V_C$, $V_{CE}$.
3. Apply a small sinusoidal input signal, for example $10\,\text{mV}$ to $20\,\text{mV}$.
4. Measure $v_{out}$ at the collector.
5. Calculate:

$$
A_v = \frac{v_{out}}{v_{in}}
$$

6. Check phase inversion and clipping.

### Common-source MOSFET amplifier

1. Set the DC bias at the gate.
2. Use $R_D$ at the drain and, if needed, $R_S$ at the source.
3. Measure the Q-point: $V_G$, $V_S$, $V_D$, $V_{DS}$.
4. Apply a small AC signal at the gate.
5. Measure the signal at the drain and calculate the gain.

## **4. Simulations with Python / Ngspice / Qucs-S**

Simulation should be used to compare three levels:

1. analytical calculation;
2. numerical simulation;
3. laboratory measurement.

The main analyses are:

* `.op` for the Q-point;
* `.dc` for $I$–$V$ characteristics;
* `.ac` for gain and phase;
* `.tran` for time response and switching.

## **5. Recommended laboratory exercises**

| Exercise | Purpose |
|---|---|
| Measuring BJT characteristics | Identifying cutoff, active operation, and saturation |
| Building a common-emitter amplifier | Analyzing Q-point and gain |
| Measuring MOSFET characteristics | Observing triode and saturation regions |
| MOSFET as a switch for an LED or small motor | Checking $R_{DS(on)}$ and power dissipation |
| SPICE circuit simulation | Comparing calculation and measurement |
| Source follower or emitter follower | Measuring input/output impedances |

#### **Final summary**

In a real transistor circuit, the analysis does not start from AC gain, but from the **Q-point**.  
Biasing places the transistor in the correct operating region, small-signal models are used only around that point, and SPICE/laboratory measurements are used for verification.
