# **1. From p–n Junction to Diode**

In the previous lecture (Lecture08), we studied how a **p–n junction** forms and how charge carriers behave at the interface.

Here, we take the next step:

> We treat the p–n junction as a **circuit element**, called a **diode**.

### **Key Idea**

A diode is the **simplest nonlinear electronic component**.

Unlike a resistor, which obeys a linear relation $I \propto V$, a diode exhibits a **strongly nonlinear current–voltage (V–I) characteristic**:

* It conducts significantly only in one direction (forward bias)
* It blocks current in the opposite direction (reverse bias), up to breakdown

</div>


### **From Physics to Device**

The microscopic processes studied in Lecture08 (diffusion, drift, depletion region, potential barrier) lead to a macroscopic behavior:

* Forward bias → barrier reduced → current flows
* Reverse bias → barrier increased → current suppressed

This behavior is what defines the **diode as a circuit element**.

# **2. V–I Characteristic of the Diode**

The behavior of a diode is fully described by its **current–voltage (V–I) characteristic**:

* Horizontal axis → voltage across the diode $V_D$
* Vertical axis → current through the diode $I_D$

The curve can be divided into three regions:


### **(a) Forward Bias Region ($V_D > 0$)**

* Current is **very small** for low voltage
* After a certain voltage (~0.6–0.7 V for silicon), current increases **very rapidly**

Key idea:
> The diode transitions from “almost OFF” to “strongly ON” in a narrow voltage range


### **(b) Reverse Bias Region ($V_D < 0$)**

* Only a **very small leakage current** flows
* The diode behaves almost like an **open circuit**

Key idea:
> Reverse current is nearly constant and very small


### **(c) Breakdown Region**

* At sufficiently large negative voltage, current increases abruptly
* This is called **breakdown**

Two cases:
* Undesired (normal diodes → damage)
* Intended (Zener diodes → voltage regulation)



## **2.2 What makes this special? (Critical Concept)**

Unlike resistors: $I = \frac{V}{R}$ a diode does **not** have a constant ratio between $V$ and $I$.

Instead:
> The diode is a **nonlinear device**: small changes in voltage can produce large changes in current.

This is why:
* linear circuit techniques **do not directly apply**
* we need **models and approximations**


## **2.3 Orders of magnitude**

Typical silicon diode:
* Forward conduction:
  * $V_D \approx 0.7\ \text{V}$ → current can be mA to A
* Reverse region:
  * current $\sim \mu$A or nA

This means:
$$
\text{Forward current} \gg \text{Reverse current}
$$

This asymmetry is what enables:
* rectification
* switching behavior



## **2.4 Dynamic perspective**

The curve is not just static — it tells you how the diode behaves in a circuit:

* Small increase in $V_D$ (forward region) → large increase in $I_D$
* Therefore:
  * diode can act like a **switch**
  * diode can strongly affect circuit operation


So far, we described the behavior **graphically**.

Next question:
> Can we describe this curve mathematically?

This leads to the **Shockley diode equation**.

# **3. Shockley Diode Equation**

The V–I characteristic of a diode can be described by the **Shockley equation**:

$$ I_D = I_S\; \left(e^{\frac{V_D}{n V_T}} - 1\right)$$


## **3.1 Meaning of the parameters**

* $I_D$ → diode current
* $V_D$ → voltage across the diode
* $I_S$ → **saturation current** (very small, typically nA–μA)
* $n$ → ideality factor (typically $1 \le n \le 2$)
* $V_T$ → **thermal voltage**

$$
V_T = \frac{kT}{q} \approx 25,\text{mV at room temperature}
$$


## **3.2 What does this equation tell us?**

This equation explains the shape you saw in the V–I curve:


### **Forward bias ($V_D > 0$)**

* Exponential term dominates:
  $$ I_D \approx I_S\; e^{\frac{V_D}{n V_T}} $$

Result:
> Current increases **exponentially** with voltage


### **Reverse bias ($V_D < 0$)**
* Exponential term becomes very small:
  $$ I_D \approx -I_S $$

Result:
> Nearly constant, very small **reverse current**


## **3.3 Key insight**

> The diode is exponential, not threshold-based.

The commonly used “0.7 V turn-on” is **not fundamental physics**—it is an **engineering approximation**.


## **3.4 Why we don’t use Shockley directly in circuits**

Although accurate, this equation is:
* nonlinear
* difficult to solve analytically in circuits

Therefore:
> In circuit analysis, we replace it with **simplified models**



## **3.5 Reverse Bias and Breakdown**

#### **Limitations of the Shockley model**

The equation is idealized and does not include:
* series resistance (important at high current)
* breakdown behavior
* temperature variation (implicitly included, but often ignored)
* real device imperfections


#### **Reverse Bias**

In the reverse bias region, the Shockley equation predicts:

$$ I_D \approx -I_S $$

This suggests a **small constant current**.
However, this is only valid up to a certain voltage.


## **3.5.1 Breakdown phenomenon**

<img src="../figs/lec7/IV_diode.png" width="30%" style="background-color:#fff;" >


When the reverse voltage exceeds a critical value:

$$ |V_D| \ge V_{BR} $$

The current increases **very rapidly**. This is called **breakdown**.


## **3.5.2 Two physical mechanisms**

There are two main mechanisms behind breakdown:


### **(a) Zener Breakdown**

* Occurs at **low breakdown voltages** (~2–6 V)
* Strong electric field causes **quantum tunneling** of electrons

Key idea:
> Electrons cross the junction without needing thermal energy


### **(b) Avalanche Breakdown**

* Occurs at **higher voltages** (> ~6–8 V)
* Carriers gain energy and **ionize atoms**, creating more carriers

Key idea:
> Chain reaction → rapid multiplication of charge carriers


## **3.5.3 Practical interpretation**

From a circuit perspective:
* Before breakdown → diode blocks current
* After breakdown → diode conducts strongly in reverse


## **3.5.4 Important distinction**

**Normal diode**
* Breakdown is **undesired**
* Can damage the device if current is not limited


**Zener diode**
* Breakdown is **intentional and controlled**
* Used for **voltage regulation**



## **3.5.5 Limitation of the Shockley equation**

> The Shockley equation does **not describe breakdown behavior**.

This is why:
* the V–I curve must be understood beyond the equation
* models must include breakdown **when relevant**


In practice:
* Breakdown is **not about voltage alone**
* It is about **power dissipation**:
$$ P = V \cdot I $$

Without current limiting:
* diode → overheats → failure



## **3.5.7 Bridge forward**

Now the full diode behavior is clear:
* Forward region → exponential conduction
* Reverse region → small leakage
* Breakdown → strong reverse conduction


## **3.6 Bridge to engineering models**

We now simplify this exponential behavior into usable approximations:
* Ideal diode model
* Threshold (constant voltage) model

These allow us to analyze circuits **without solving exponential equations**.

# **4. Engineering Models of the Diode**

The Shockley equation accurately describes the diode, but it is **not practical** for most circuit analysis.

Therefore:
> We replace the exponential behavior with **simplified models** that capture the essential features.


## **4.1 Why models are needed**

From the previous section:

* Forward region → exponential growth
* Reverse region → almost zero current

This suggests a simpler approximation:

> The diode behaves almost like a **switch controlled by voltage**.



## **4.2 Ideal Diode Model**

### **Definition**

* Forward bias → diode conducts perfectly
* Reverse bias → diode blocks completely

$$ \text{Forward: } V_D = 0,\quad I_D > 0$$
$$ \text{Reverse: } I_D = 0,\quad V_D < 0 $$

* Acts like a **closed switch** in forward bias
* Acts like an **open switch** in reverse bias

## **4.3 Threshold (Constant Voltage) Model**

### **Definition**

<img src="../figs/lec7/Diode_Modelling.png" width="30%" style="background-color:#fff;" >


* Diode turns ON at a fixed voltage:
  $$ V_D \approx V_t $$

Typical values:
* Silicon diode → $V_t \approx 0.7,\text{V}$
* Germanium diode → $V_t \approx 0.3,\text{V}$

**Behavior**
* No current below threshold
* After threshold → voltage ~ constant, current increases freely

**Interpretation**
> The exponential curve is approximated by a **vertical shift**.


**Use case**
* Most practical circuit analysis
* Good balance between simplicity and accuracy

## **4.4 Piecewise-linear refinement**

<img src="../figs/lec7/Diode_Modelling_Real.png" width="30%" style="background-color:#fff;" >

<img src="../figs/lec7/Rd_model.svg" width="50%" style="background-color:#fff;" >

We can improve the model slightly:
$$ V_D = V_t + I_D\times r_d $$

where: $r_d$ → small dynamic resistance

**Interpretation**
* Captures:
  * threshold behavior
  * slight slope in forward region




## **4.5 Model hierarchy**

Notice:

| Model            | Accuracy | Complexity  | Use                   |
| ---------------- | -------- | ----------- | --------------------- |
| Ideal            | Low      | Very simple | Conceptual            |
| Threshold        | Medium   | Simple      | Most circuits         |
| Piecewise-linear | Higher   | Moderate    | More precise analysis |
| Shockley         | High     | Complex     | Simulation / theory   |




## **4.6 Key takeaway**

> All diode models are approximations of the same exponential law.
> * The **physics does not change**
> * Only the **level of approximation changes**



## **4.7 Bridge to applications**

Now we can use these models to analyze circuits:

* Rectifiers
* Clippers
* Clampers

Without solving exponential equations.

# **5. Types of Diodes**

While all diodes are based on the same **p–n junction principle**, their **construction is optimized** for different applications.

Therefore, different diodes exhibit different:
* voltage ranges
* speed
* breakdown behavior
* light emission


## **5.1 Common diode types**

| Type                           | Typical Example | Key Feature                     | Typical Use                       |
| ------------------------------ | --------------- | ------------------------------- | --------------------------------- |
| **Small-signal diode**         | 1N4148          | Fast switching, low current     | Signal processing, logic circuits |
| **Rectifier diode**            | 1N4007          | Handles high current/voltage    | Power supplies (AC → DC)          |
| **Zener diode**                | BZX55, 1N4733   | Controlled breakdown voltage    | Voltage regulation                |
| **LED (Light Emitting Diode)** | Various         | Emits light when forward biased | Indicators, displays, lighting    |



## **5.2 Key distinctions**

Note: What actually metters at this stage are not the names but their principles.

#### **(a) Forward voltage $V_t$ (threshold)**
* Silicon → ~0.7 V
* LED → 1.8–3.3 V (depends on color/material)

#### **(b) Maximum current**

* Small-signal → mA
* Rectifier → A range


#### **(c) Reverse breakdown voltage**

* Standard diode → must **not be reached**
* Zener diode → **designed to operate here**


#### **(d) Switching speed**

* Important in high-frequency circuits
* Fast diodes (e.g., 1N4148) vs slow rectifiers (e.g., 1N4007)



## **5.3 Important conceptual clarification**

> The *type* of diode does not change the fundamental behavior.

All diodes still follow:
* exponential I–V characteristic
* same modeling approaches

What changes is:
* parameter values
* operating region


## **5.4 Practical selection mindset**

When choosing a diode, you should ask:
* What current must it handle?
* What reverse voltage must it withstand?
* Does switching speed matter?
* Is light emission required?


## **5.5 Bridge to applications**

Now that we understand:

* how a diode behaves (V–I curve)
* how to model it (ideal, threshold)
* what types exist

we can use it in circuits, e.g.:
* rectification
* clipping
* voltage shaping


# **6. Rectification: Converting AC to DC**

One of the most important uses of a diode is **rectification**:
> Converting an alternating signal (AC) into a unidirectional signal (DC).


## **6.1 Half-Wave Rectifier**

### **Circuit**

<img src="../figs/lec7/rectifier_halfWave.png" width="50%" style="background-color:#fff;" >

### **Operation (using threshold model)**

Assume:
* Silicon diode → $V_t \approx 0.7,\text{V}$
* Input: $V_{in}(t) = V_0 \sin(\omega t)$


**Case 1: Positive half-cycle** $ V_{in} > V_t$:
* Diode is **ON**
* Output: $ V_{out} \approx V_{in} - V_t $

The signal passes, but slightly reduced due to $V_t$.


**Case 2: Negative half-cycle** $ V_{in} < 0 $
* Diode is **OFF**
* Output: $ V_{out} = 0 $

Signal is blocked.


**Result**
> The output contains only the **positive cycles** of the input.


## **6.2 Key insight**

The diode acts as a **polarity-dependent switch**:
* ON → passes signal
* OFF → blocks signal

This is the foundation of:
* power supplies
* signal shaping


## **6.3 Waveform interpretation**

You must learn to **read circuits in time**, not just statically:
* Input: sinusoidal
* Output: **clipped waveform**

Important:
* Output is **not constant DC**
* It is **pulsating DC**!



## **6.4 Limitations of half-wave rectifier**

* Only uses half of the signal
* Large ripple (variation in output)
* **Inefficient!**

### **Circuit**

# **7. Clipper Circuits**

A **clipper** is a circuit that **limits (clips) the voltage** to a specified level.

> It removes parts of the signal that exceed a certain threshold.


## **7.1 Basic Idea**

* When the diode is **OFF** → signal passes unchanged
* When the diode is **ON** → voltage is **limited**

So the analysis is always:
> Determine when the diode is ON or OFF


## **7.2 Positive Clipper**

### **Circuit**

<img src="../figs/lec7/clipper.svg" width="30%" style="background-color:#fff;" >

### **Operation (threshold model)**

Assume: $V_t \approx 0.7,\text{V}$

**Case 1: $V_{in} < V_t$**
* Diode is **OFF**
* No current flows through diode
$ V_{out} \approx V_{in} $
Signal passes unchanged


**Case 2: $V_{in} \ge V_t$**

* Diode is **ON**
* Output is clamped:
$V_{out} \approx V_t $

Signal is **clipped at ~0.7 V**

**Result**
> The output cannot exceed the diode threshold voltage.


## **7.3 Negative Clipper**

**Concept**
By reversing the diode:
* Positive part → unchanged
* Negative part → clipped

**Behavior**
* For negative input below $-V_t$ → diode ON
* Output limited near $-V_t$



## **7.4 Biased Clipper (adjustable threshold)**

### **Circuit idea**

Add a DC source $V_B$:
* Clipping occurs at:
  $ V_{clip} \approx V_B + V_t $

## **7.5 General Analysis Strategy**

When analysing circuit with diodes we must follow:

* **Step 1: Assume diode state**:
    * ON or OFF
* **Step 2: Replace diode with model**
  * ON → $V_D = V_t$
  * OFF → open circuit
* **Step 3: Solve circuit**
* **Step 4: Check consistency**

Does the result match the assumption?

Actually, this is a **standard method for all nonlinear circuits**!


## **7.6 Key insight**

> Clippers do not change the entire signal — only the parts beyond a threshold.


## **7.7 Practical applications**

* Signal protection (limit voltage)
* Waveform shaping
* Analog signal conditioning


## **7.8 Bridge to next section**

Clippers **limit amplitude**.

Next:
> Clampers **shift the entire signal level**

# **8. Clamper Circuits**

A **clamper** is a circuit that **shifts the entire signal up or down** without changing its shape.

> Unlike clippers, clampers do not remove parts of the signal—they **add a DC offset**.


## **8.1 Basic Idea**
Uses:
* diode
* capacitor
* resistor

The capacitor stores charge and **redefines the reference level** of the signal.

## **8.3 Operation**

Assume sinusoidal input.

* **Step 1: Negative half-cycle**
    * Diode becomes **ON**
    * Capacitor charges approximately to:  $V_C \approx V_{peak} - V_t$
Capacitor stores voltage

* **Step 2: Positive half-cycle**
    * Diode is **OFF**
    * Capacitor acts like a **voltage source in series** $V_{out} = V_{in} + V_C$
Entire waveform shifts **upward**


## **8.4 Result**

> The signal is shifted so that its minimum value is approximately $-V_t$ (or near 0 depending on approximation).


## **8.5 Negative Clamper**

By reversing the diode:
* Capacitor charges during positive half-cycle
* Signal shifts **downward**


## **8.6 Key condition**

For proper operation: 
$$ \tau=RC \gg T $$

where:
* $T$ = period of input signal

This ensures:
* capacitor does **not discharge significantly** between cycles


## **8.7 Key insight**

Remember:

> Clippers modify amplitude
> Clampers modify reference level

Don't confuse them!


## **8.8 Practical applications**

* DC level shifting
* Signal conditioning before amplification
* Television and communication circuits (historically important)


## **8.9 Connection to previous concepts**

Clamper operation depends on:
* diode switching (ON/OFF)
* capacitor charge storage
* time constants

This connects back to:
* RC circuits
* transient behavior
