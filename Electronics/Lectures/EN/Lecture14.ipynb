## Summary:

Time measurement and the generation of periodic signals are fundamental in analog and digital electronics. We have two main categories of circuits:
* **Oscillators**: Generate periodic signals such as sinusoidal, triangular, or square waves, etc., without requiring a periodic input signal (astable operation).
* **Timers**: Are closely related to oscillators. They are circuits that produce a delayed pulse or a pulse with a defined time width after a triggering event (monostable operation). For timers we usually speak about time delay or pulse width, not frequency as in the case of oscillators.

These circuits are used in clocks, signaling, communication, processors, PWM, etc.


#### **Principle of operation**

* Oscillation comes from the **charging/discharging or exchange of energy** in capacitors and/or inductors.
* **LC circuit**: energy alternates between the electric and magnetic fields.
* Capacitors and resistors are used for **time control** (RC time constant).
* **Positive feedback + amplification** → create the conditions for oscillation (Barkhausen criterion).


#### **Classification of oscillators**
* **According to the output waveform:**
  * Sinusoidal (Wien Bridge)
  * Square wave (555 astable)
  * Triangular wave, noise, etc.
* **According to the circuit components:**
  * **RC oscillators** – for low frequencies
  * **LC oscillators** – for RF signals
  * **Crystal oscillators** – for very accurate frequency
  * **Relaxation oscillators** – based on capacitor charging/discharging
* **Analog vs digital oscillators**


#### **555 timer IC and the astable circuit**

The 555 timer IC is one of the best-known and most widely used integrated circuits, especially in education, for applications such as timers, delay circuits, pulse generators, and oscillators. The 555 has three main operating modes, and it can also be used as a Schmitt trigger in special configurations:
1. **Astable mode** (free running): Works as an oscillator and generates square-wave signals without requiring an input signal. Applications: flashing lamps and LEDs, pulse generation, pulse-width modulation (PWM), logic clock signals, tone generation, security alarms (sirens), etc.

2. **Monostable mode** (one-shot): Works as a one-shot generator. Applications: timers, missing-pulse detection, switch debouncing, touch switches, frequency dividers, and triggered measurement of resistance or capacitance.

3. **Bistable mode** (latch): Works as a set-reset latch: it stores one of two states until it is changed by an external signal. A typical application is switch debouncing.

4. Use as a **Schmitt trigger** (inverter): Can be used as an inverting Schmitt trigger. Application: converts a noisy or slowly varying input into a clean digital signal.



#### **Analog oscillators**
* **Wien Bridge** – clean sinusoidal signal, audio frequency range.
* **RC Phase Shift** – requires three RC stages for 180° phase shift.
* **Hartley & Colpitts** – use an LC circuit for RF signals.
* Used in audio, communication, transmitters, and signal generators.


#### **Digital oscillators**
* **Ring oscillator** – inverters connected in a loop; simple square-wave signal.
* **Schmitt-trigger oscillator** – uses hysteresis for a stable waveform.
* **PLL (Phase Locked Loop)** – for phase/frequency synchronization with a reference signal.
* All are used in **processors, FPGAs, clock signals, digital communication**, etc.

## **Why do we need oscillators and timers in electronics?**

In electronics, **time** is a fundamental quantity. We not only want to measure time, but also to **generate periodic signals** such as digital pulses or sinusoidal waves.

The two key terms we will encounter are:

* **Timer** – a circuit that “counts time” and reacts after a specified delay.
* **Oscillator** – a circuit that **generates a periodic signal**, usually a voltage that changes with time.

## **Main difference:**

| Circuit type   | Main function                              | Example                                 |
| -------------- | ------------------------------------------- | --------------------------------------- |
| **Timer**      | Gives a pulse or action after a delay       | Turn on an LED after 2 seconds          |
| **Oscillator** | Generates a periodic signal continuously    | Digital clock, PWM, music               |

## **Real-life examples:**

* Digital clocks → use **crystal oscillators** to count seconds.
* Microwave ovens → use a **timer** to stop heating after a set time.
* Signal communication (Wi-Fi, Bluetooth) → uses **oscillators** to create synchronized signals.

## **What is an oscillator?**

An **oscillator** is a system that generates periodic signals without external periodic excitation, that is, without needing a periodic input signal. In general, the circuit takes energy from the power supply and, through feedback and frequency- or time-setting elements, produces a signal that varies periodically in time. For sinusoidal LC/RC oscillators, this process is closely related to resonance; for relaxation oscillators, it is mainly related to the periodic charging and discharging of a capacitor:

> **Energy is stored and exchanged periodically, or a capacitor is charged and discharged between defined threshold levels.**

Example:
* In an **LC circuit**, energy alternates between:
  * **L (inductor)**, which stores energy in the magnetic field,
  * **C (capacitor)**, which stores energy in the electric field.

This combination **can create a natural sinusoidal signal** with a frequency that depends on L and C (the resonance frequency):
$$
f_0 = \frac{1}{2\pi \sqrt{LC}}
$$

#### **Role of passive and active components**

| Component                                    | Role in the oscillator                              |
| -------------------------------------------- | --------------------------------------------------- |
| **Resistor (R)**                             | Controls the charging/discharging time in RC timers |
| **Capacitor (C)**                            | Stores energy in electric form                      |
| **Inductor (L)**                             | Stores energy in magnetic form                      |
| **Active component (transistor, op-amp, IC)**| Provides amplification and creates **positive feedback** |


#### **Positive feedback and the Barkhausen criterion**

In most electronic oscillators, the signal is sustained by a **positive feedback mechanism**: part of the output signal is returned to the input with the proper phase so that losses are compensated.

#### **Conditions for oscillation (Barkhausen):**

1. **Net positive feedback** – The returned signal must be **in phase with the input** at the oscillation frequency.
2. **Loop gain** – To start oscillation, usually $|A\beta| > 1$ is needed; for sustained oscillation, nonlinearities or amplitude control effectively bring it to $|A\beta| \approx 1$.

If these conditions are satisfied and the amplitude is physically limited, the circuit can enter a stable oscillatory state.

### **Timers and capacitor charging**

In the simplest **RC timer** circuit, the **delay time** depends on how fast a capacitor charges through a resistor:
$$
V(t) = V_{max} \left( 1 - e^{-t/RC} \right)
$$

This is the core idea behind many **timers**, including the famous **555 timer IC**.

## **Classification of oscillators**

Oscillators are classified in several ways, depending on:

* **The output waveform**
* **The components used to set the frequency**
* **The circuit architecture and type of feedback**

### **According to waveform**

| Waveform type      | Description                              | Applications                  |
| ------------------ | ---------------------------------------- | ----------------------------- |
| **Sinusoidal**     | Smooth wave, without strong harmonics    | Communication, clocks         |
| **Square**         | Fast switching between two levels        | Digital systems               |
| **Triangular**     | Linear rise and fall of voltage          | PWM control, music            |
| **Noise**          | Random signal                            | Testing, biological simulation |


### **According to architecture and dominant components**

#### **RC oscillators**

* Use **resistors and capacitors** to set the frequency.
* Usually used for **low-frequency signals** (Hz to several kHz).
* **Examples:**

  * **Phase-shift oscillator**
  * **Wien Bridge oscillator**

* Use an **inductor (L)** and **capacitor (C)** for resonance.
* More suitable for higher frequencies (RF, MHz).
* **Examples:**

  * **Colpitts**
  * **Hartley**

## **Timers and the astable circuit – 555 timer IC**

### **What is the 555 timer IC?**

The 555 timer IC is a very widely used **integrated circuit** (IC) that can work in three modes:

| Operating mode | Description                                                                                     |
| -------------- | ----------------------------------------------------------------------------------------------- |
| **Monostable** | Produces one pulse after a triggering signal                                                     |
| **Astable**    | Continuously generates square waves (oscillator)                                                 |
| **Bistable**   | Works like a flip-flop: remains in one of two states until changed by an external signal         |


### **Internal structure of the 555 timer IC (briefly)**

Internally, the 555 consists of:

* 2 **comparators** that compare input voltages with the levels 1/3 and 2/3 of VCC
* 1 **RS flip-flop** that stores the logic state
* 1 **discharge transistor**
* Several **resistors** for voltage division


<img src="../figs/lec10/NE555_Bloc_Diagram.svg" width="50%" style="background-color:#fff;" >


### **Astable circuit with the 555 timer IC: a simple oscillator**

In this configuration, the 555 timer IC **continuously generates square waves**, without needing an input signal. This is a **relaxation oscillator**, which works by charging and discharging a capacitor.


#### **Formulas for frequency and duty cycle**

$$
T = 0.693 (R_1 + 2R_2)C
\quad\text{(total period)}
$$

$$
f = \frac{1}{T} = \frac{1.44}{(R_1 + 2R_2)C}
$$

$$
\text{Duty Cycle} = \frac{R_1 + R_2}{R_1 + 2R_2}
$$

This oscillator is well suited for **pulse and PWM signal generation**, simple signaling waveforms, and LED testing.

<img src="../figs/lec10/555_Astable_Diagram.svg" width="40%" style="background-color:#fff;" > <img src="../figs/lec10/NE555_Astable_Waveforms.svg" width="40%" style="background-color:#fff;" >


### **Monostable circuit: one-pulse timer**

In this configuration, the 555 timer IC is at **rest** until a triggering signal activates it and makes it produce **one pulse with a duration set by R and C**.

### Schematic idea:

* The button provides a negative trigger
* The output voltage (OUT) goes high for the time:

$$
T = 1.1 \cdot R \cdot C
$$

<img src="../figs/lec10/555_Monostable.svg" width="40%" style="background-color:#fff;" > <img src="../figs/lec10/NE555_Monotable_Waveforms.png" width="40%" style="background-color:#fff;" >




### **Summary:**

* The 555 timer IC is an **ideal basis for experiments**: very cheap and very flexible.
* It can be used as an **oscillator** (generates signals) or as a **timer** (acts after a delay).
* The values of **R and C** determine the response time or frequency.
* The **astable circuit** has no stable state – the waves continue without stopping.
* The **monostable circuit** has one stable state – it reacts only when “triggered”.

## **Analog oscillators**

Here we consider **sinusoidal oscillators**, which produce **continuous and spectrally clean waves**, suitable for audio, radio, signaling, and measurement systems. These oscillators use **passive components (R, C, L)** together with **active components (transistors or op-amps)** to provide **positive feedback and amplification**.


### **Basic principle of sinusoidal oscillators:**

> **Periodic storage and exchange of energy** + **positive feedback** + **controlled amplification**.

### **Wien Bridge oscillator**

One of the most famous oscillators for **clean sinusoidal signals** at low frequencies – widely used for audio experiments.

#### Main components:

* Two resistors and two capacitors in an **RC filter network**
* An **op-amp** that provides amplification
* An **amplitude-control mechanism** to keep the signal stable.


<img src="../figs/lec10/960px-Wien_Bridge_Oscillator.png" width="50%" style="background-color:#fff;" >


### **RC phase-shift oscillator**

Uses several **RC filters** to create a total phase shift of **180°**, combined with **180° from an amplifier**, satisfying the Barkhausen criterion for positive feedback.

#### **Oscillation frequency**:

$$
f_0 = \frac{1}{2\pi RC\sqrt{6}} \quad (\text{with 3 RC stages})
$$

#### Used for:

* Low frequencies
* Audio devices
* Sinusoidal test signals

<img src="../figs/lec10/RC_phase_shift_oscillator.svg" width="50%" style="background-color:#fff;" >



### **Hartley oscillator**

This oscillator uses an **LC circuit** with a **tapped inductor** for feedback. It is ideal for **radio frequencies**.

$$
f_0 = \frac{1}{2\pi \sqrt{L_{eq}C}} \quad \text{with } L_{eq} = L_1 + L_2
$$

#### Typical use:

* RF transmitters and receivers
* Resonant detectors


<img src="../figs/lec10/Oscillator_comparison.svg" width="50%" style="background-color:#fff;" >


### **Colpitts oscillator**

Similar to the Hartley oscillator, but the **feedback is taken from a capacitive divider**, while the inductor is single.

$$
f_0 = \frac{1}{2\pi \sqrt{L \cdot C_{eq}}}, \quad C_{eq} = \frac{C_1 C_2}{C_1 + C_2}
$$

#### Used for:

* RF and microwave signals
* Stable resonant circuits



### **Quick comparison**

| Oscillator type | Frequency       | Dominant components   | Applications      |
| --------------- | --------------- | --------------------- | ----------------- |
| Wien Bridge     | low (audio)     | RC + op-amp           | Sine generator    |
| Phase Shift     | low – medium    | RC + transistor/op-amp| Test signal       |
| Hartley         | RF              | L + C + transistor    | Transmitter       |
| Colpitts        | RF/microwave    | L + C + transistor    | Stable RF         |


* Analog oscillators are based on **resonance** and **positive feedback**.
* RC oscillators are simple but frequency-limited.
* LC oscillators reach high frequencies and are more stable.
* Stability and amplitude must be carefully controlled to avoid decay or unlimited growth of the signal.






## **Digital oscillators and simple circuits**

### **What is a digital oscillator?**

A **digital oscillator** is a circuit that **generates a periodic digital signal**, usually a **square wave**, used for:

* Digital systems (processors, microcontrollers)
* Timing generation (clock)
* PWM control, signaling, UART, SPI, etc.


### **Characteristics of digital oscillators:**

* Produce signals with **clear logic levels** (0 V or VCC)
* Usually very simple to build
* Do not require analog components such as L or C in some cases
* Frequency is determined by **the use of counters, delays, or resonators**


### **Ring oscillator**

#### Description:

A **closed loop** made from an odd number of **logic inverters**, where each inverter introduces a small delay, creating a periodic waveform.

<img src="../figs/lec10/Ring_oscillator.svg" width="50%" style="background-color:#fff;" >

$$
f = \frac{1}{2N \cdot t_d}
$$

where:

* $N$ = number of inverters (must be **odd**)
* $t_d$ = delay time of one inverter

#### Used in:

* FPGAs
* Processors
* Delay measurements



### **Schmitt-trigger oscillator**

Uses a **Schmitt-trigger inverter** and an **RC network** to create a very simple and stable oscillator.

<img src="../figs/lec10/Schmitt_trigger_symbol.svg" width="30%" style="background-color:#fff;" > <img src="../figs/lec10/Op-Amp_Schmitt_Trigger.svg" width="50%" style="background-color:#fff;" >


The capacitor charges and discharges through the resistor, while the inverter switches the signal when the voltage crosses the **trigger thresholds** (hysteresis).



### **Phase-Locked Loop (PLL)**

#### Description:

A system that **synchronizes the frequency and phase of an internal oscillator with a reference signal**.

#### Components:

* **Phase Detector** – Compares the input signal with the internal signal
* **Low Pass Filter** – Filters the phase-difference signal
* **VCO (Voltage Controlled Oscillator)** – Changes the frequency according to the control voltage
* **Feedback Loop** – Creates closed-loop control

$$
f_{out} = N \cdot f_{ref}
$$

(if a divider is used in the feedback path)

#### Used for:

* Synchronization in telecommunications (FM, AM)
* Stable frequency generation in microchips
* Digital modulation/demodulation



### **Comparison: analog vs digital oscillators**

| Type                          | Signal type     | Stability           | Simplicity           | Applications               |
| ----------------------------- | --------------- | ------------------- | -------------------- | -------------------------- |
| Analog (RC/LC)                | Sinusoidal      | High with crystal   | Medium – complex     | Audio, RF                  |
| Digital (Ring, Schmitt, PLL)  | Square/logic    | High with PLL/crystal| Simple               | Digital systems, clocks    |

* Digital oscillators are **building blocks of computer systems**
