# **I. Introduction**

## **What is a transistor?**

A transistor is a semiconductor electronic device that can behave either as an electrically controlled switch or as an electrically controlled amplifier. At the core of transistor operation is the control of current flow, analogous to controlling water flow with a valve. Unlike the passive devices we have studied so far, which had only two terminals, transistors have an additional control terminal; therefore, they are three-terminal devices. A small voltage or current signal applied to the control terminal can produce a large change in current through the other two terminals. Because of their wide range of applications, transistors are found in practically every electronic circuit in modern devices.


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Definition**

**A transistor is a three-terminal semiconductor electronic device that controls the flow of current in a circuit through an external control signal, either a current or a voltage.**

- If the control is by current through carrier injection ➔ the device is a bipolar junction transistor, abbreviated **BJT**.
- If the control is by voltage through electrostatic control of a channel ➔ the device is a field-effect transistor, abbreviated **FET**.

</div>

- A transistor controls current with a small electrical signal, either current or voltage, and **amplifies** the effect of this signal into a much larger output signal. Therefore, **the transistor is an amplifier**.
- For sufficiently large signals, **the transistor enters saturation** and effectively becomes an **electrically controlled switch**.
- It is the **heart** of modern electronics: microprocessors, sensors, digital communication, and many other systems all depend on transistors.


<br>

**In essence**, a transistor is an **electrical valve**: it uses a **small signal** to control a **larger signal**.

## Historical overview of the transistor

- **1925:** Lilienfeld imagined a device whose electrical resistance could be controlled by an electric field, but the available materials were not pure enough.
- In the 1940s, technology relied on **vacuum tubes** for amplification and switching. These were large, consumed a lot of power, and were not very reliable. A **small, reliable, and energy-efficient replacement** was needed.
- **1947:** The first practical transistor was invented: the BJT, by Bardeen, Brattain, and Shockley at Bell Labs.
- **1959:** After problems with oxides, especially SiO₂, were solved, the construction of the MOSFET became possible.
- Today: the MOSFET dominates digital electronics and integrated circuits (ICs), while BJTs are used mainly in amplifiers and high-current analog circuits.

<img src="../figs/lec8/Triode_symbol.svg" alt="Triode" width="44%" style="background-color:#fff;" > <img src="../figs/lec8/Triode.svg" alt="Triode" width="45%" style="background-color:#fff;" >


**Historical irony:**  
Although the BJT transistor (1947) was discovered before the practical FET/MOSFET (1959), the first conceptual idea was actually for a **field-effect transistor (FET)** in 1925. At that time, however, it was not possible because of material imperfections.

# **II.A — Structure and Physical Operation of the Bipolar Junction Transistor (BJT)**

## **1. Basic BJT structure**

<img src="../figs/lec8/npn_Cross-section.svg" alt="npn cross section" width="30%" style="background-color:#fff;" > <img src="../figs/lec8/BJT_pnp_npn.svg" alt="bjt" width="50%" style="background-color:#fff;" >

A **bipolar junction transistor** (BJT) consists of **three semiconductor regions**:
* **Emitter (E)** — heavily doped in order to supply many carriers.
* **Base (B)** — thin and lightly doped, allowing only a small amount of recombination.
* **Collector (C)** — less heavily doped and wider, designed to collect carriers.

BJTs are divided into two types:
* **NPN**: n–p–n
* **PNP**: p–n–p

In the following, we focus on the **NPN** type because it is more commonly used and more intuitive for analysis. The analysis for the PNP transistor is identical, except that the voltage polarities and current directions are reversed. Everything else in the analysis is the same.


## **2. Carrier injection and transport**

In an **NPN** transistor operating in the **forward active region**:
* the **emitter–base** junction is **forward-biased**;
* the **base–collector** junction is **reverse-biased**.

This produces the following sequence of events:
1. Electrons are injected from the **emitter** (n) into the **base** (p).
2. The base is **very thin**, so **most electrons do not recombine**; instead, they pass through the base.
3. On the other side, the **electric field of the base–collector junction** pulls these electrons toward the **collector**.

<img src="../figs/lec8/npn_Operation.svg" alt="npn cross section" width="80%" style="background-color:#fff;" >

<img src="../figs/lec8/NPN_Band_Diagram_Active.svg" alt="npn cross section" width="50%" style="background-color:#fff;" >


## **3. Currents through the three terminals**

In this process, three main currents appear:
* **$I_E$**: emitter current — flows out of the emitter.
* **$I_B$**: base current — very small, approximately 1% of $I_E$.
* **$I_C$**: collector current — almost equal to $I_E$.

The currents satisfy:
$$ I_E = I_B + I_C $$

The collector current is:
$$ I_C = \beta I_B $$
where $\beta$ is the **current gain**, usually $50 \leq \beta \leq 300$.


## **4. Physical origin of amplification**

Amplification occurs because:
* a **very small current** in the base ($I_B$)
* **controls the flow of a much larger current** in the collector ($I_C$).

This ratio creates the effect of **current amplification**.


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary of structure and function**

* The BJT uses **carrier injection** and an **electric field** to control current flow.
* The **asymmetric structure** — thin base and heavily doped emitter — is the key to efficient amplification.
* This physical operation makes it possible to model the device either as an amplifier or as a switch.

</div>

#### **Main physical mechanism of the BJT**

In the forward active region:

1. The **emitter–base** junction is forward-biased.
2. The emitter injects many carriers into the base.
3. The base is very thin and lightly doped, so only a few carriers recombine.
4. Most carriers reach the base–collector junction.
5. The electric field in the base–collector junction pulls the carriers toward the collector.

Therefore, a small change in $V_{BE}$ or $I_B$ produces a much larger change in $I_C$.

</div>


# **II.B — Operating Regions of the Bipolar Junction Transistor (BJT)**

The bipolar transistor (BJT) has **four main operating regions**, determined by the **biasing of the two p-n junctions**:
* the **emitter–base junction (BE)**;
* the **base–collector junction (BC)**.

These regions are important for understanding **how the BJT is used as an amplifier or as a switch**.

<img src="../figs/lec8/bjt_char.png" alt="npn cross section" width="50%" style="background-color:#fff;" >


## **Summary table of operating regions**


| Region | Emitter–base junction (BE) | Base–collector junction (BC) | Main use |
|---|---|---|---|
| **Cutoff** | reverse-biased | reverse-biased | open switch, $I_C \approx 0$ |
| **Forward active** | forward-biased | reverse-biased | linear amplification |
| **Saturation** | forward-biased | forward-biased | closed switch |
| **Reverse active** | reverse-biased | forward-biased | rarely used |



## **1. Forward active region**

* **BE** is **forward-biased**:
  $$ V_{BE} \gtrsim 0.6\ \text{V (for silicon)} $$
* **BC** is **reverse-biased**:
  $$ V_{CB} > 0 $$

**Result:**
* Electrons are injected from the emitter into the base.
* The collector attracts them → $I_C \approx \beta I_B$ flows.
* This is the **desired region for amplification**.

## **2. Saturation region**

* Both junctions, **BE and BC**, are **forward-biased**.
  $$ V_{BE} \gtrsim 0.6\, \text{V},\quad V_{CB} < 0 $$

**Result:**
* The base region is “filled with carriers.”
* The transistor behaves as a **closed switch**.
* The current $I_C$ is limited by the external circuit → it no longer follows $I_C = \beta I_B$.


## **3. Cutoff region**

* Both junctions are **reverse-biased**:
  $$ V_{BE} < 0,\quad V_{CB} < 0 $$

**Result:**
* No carriers are injected.
* The current $I_C \approx 0$ → **the transistor is off**.
* Used as an **OFF switch** in digital logic.

## **4. Reverse active region**

* **BE** is **reverse-biased** and **BC** is **forward-biased**.
* The structure is reversed → the collector acts as the emitter.

**Result:**
* It works, but **with much lower efficiency**:
  * $\beta_{\text{reverse}} \ll \beta_{\text{forward}}$
* It is rarely used → mostly in theoretical analysis or during switching transitions.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary of operating regions**

* The **forward active** region is the **heart of analog amplification**.
* The **saturation** and **cutoff** regions are fundamental in **digital logic** (ON/OFF).
* Understanding the biasing condition of each junction is essential for circuit design.

</div>

#### **Terminology warning: saturation in a BJT**

For a **BJT**, saturation means that both p-n junctions are forward-biased. In this region, the transistor no longer behaves as a linear amplifier, and the relation $I_C = \beta I_B$ is no longer valid.

In practice, BJT saturation is usually used when the transistor is operated as a **closed switch**.

</div>


<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

#### **Practical hierarchy of BJT models**

1. **Simplified DC model**  
   Used for the operating point: $V_{BE} \approx 0.7\,\text{V}$ and $I_C \approx \beta I_B$.

2. **Exponential model**  
   Used to understand the physical dependence in the active region:
   $$
   I_C = I_S e^{V_{BE}/V_T}
   $$

3. **Ebers–Moll model**  
   A more complete model for all BJT operating regions.

4. **Hybrid-$\pi$ model**  
   Used for small-signal analysis after the operating point has been found.

</div>


# **II.C — Mathematical Modeling of the BJT: Ebers–Moll Model and Approximations**

## **Purpose of modeling**

The mathematical model of the bipolar transistor is used to describe **the currents through its terminals** ($I_E$, $I_B$, $I_C$) as functions of **the applied voltages** and **the physical properties** of the device.

The **Ebers–Moll** model is one of the most important BJT models because:

* It describes **all operating regions** of the BJT.
* It is based on **carrier transport** and **p-n junction behavior**.
* It uses the concepts of two **saturation currents** and two **gain factors**.


## **1. Full equations of the Ebers–Moll model**

<img src="../figs/lec8/Ebers-Moll_model_schematic_(NPN).svg" alt="npn cross section" width="40%" style="background-color:#fff;" > <img src="../figs/lec8/Ebers-Moll_model_schematic_(PNP).svg" alt="npn cross section" width="40%" style="background-color:#fff;" >

For an **NPN** transistor, the currents are described as follows:

$$
\begin{aligned}
I_E &= I_{ES} \left( e^{\frac{V_{BE}}{V_T}} - 1 \right) - \alpha_R I_{CS} \left( e^{\frac{V_{BC}}{V_T}} - 1 \right) \\
I_C &= \alpha_F I_{ES} \left( e^{\frac{V_{BE}}{V_T}} - 1 \right) - I_{CS} \left( e^{\frac{V_{BC}}{V_T}} - 1 \right) \\
I_B &= I_E - I_C
\end{aligned}
$$

**Symbols and physical meanings**

| Symbol | Meaning |
| ------------- | ----------------------------------------------------------------- |
| $I_{ES}$ | Saturation current of the emitter–base junction |
| $I_{CS}$ | Saturation current of the base–collector junction |
| $V_{BE}$ | Voltage between base and emitter |
| $V_{BC}$ | Voltage between base and collector |
| $V_T$ | Thermal voltage, $V_T = \frac{kT}{q} \approx 25.85\, \text{mV}$ |
| $\alpha_F$ | Forward carrier gain, usually $0.98 - 0.998$ |
| $\alpha_R$ | Reverse gain, usually much smaller: $0.2 - 0.5$ |


## **2. Approximations in the forward active region**

<img src="../figs/lec8/Approximated_Ebers_Moll.svg" alt="npn cross section" width="40%" style="background-color:#fff;" >

When the BJT is in the **forward active region**:
* $V_{BE} \gg V_T$ → $e^{\frac{V_{BE}}{V_T}} \gg 1$
* $V_{BC}$ is negative (reverse bias) → $e^{\frac{V_{BC}}{V_T}} \approx 0$

### The simplifications become:
$$
\begin{aligned}
I_C &\approx \alpha_F I_{ES} e^{\frac{V_{BE}}{V_T}} \\
I_E &\approx I_{ES} e^{\frac{V_{BE}}{V_T}} \\
I_B &\approx I_E - I_C = (1 - \alpha_F) I_{ES} e^{\frac{V_{BE}}{V_T}}
\end{aligned}
$$


### Relation with the parameter $\beta$

It is often preferable to use:
$$
\beta = \frac{\alpha_F}{1 - \alpha_F}
\quad \text{and} \quad
\alpha_F = \frac{\beta}{1 + \beta}
$$

Using these relations, we obtain:
$$
I_C = \beta I_B \quad \text{and} \quad I_E = (1 + \beta) I_B
$$


## **3. Early effect and improved model**

When $V_{CE}$ increases in the active region, the effective base width decreases → $I_C$ increases slightly more than predicted by the simple exponential model. From the expression $I_C = I_S\,\exp\left(\frac{V_{BE}}{V_T}\right)$, we see that for a fixed value $V_{BE}=\text{const}$, $I_C$ would be independent of $V_{CE}$. In a real transistor, when $V_{CE}$ increases, the base effectively becomes thinner because of the Early effect, and the current increases, as shown in the figure below.

<img src="../figs/lec8/Early_effect_(NPN).svg" alt="npn cross section" width="40%" style="background-color:#fff;" >


### The corrected expression is:
$$
I_C = I_S e^{\frac{V_{BE}}{V_T}} \left(1 + \frac{V_{CE}}{V_A}\right)
$$
where:
* $V_A$ is the **Early voltage**, usually $50 - 150$ V.
* This effect is important in **small-signal models**, where it determines the output resistance:
$$
r_o \approx \frac{V_A}{I_C}
$$


<img src="../figs/lec8/Early_effect_ICvsVCE).svg" alt="npn cross section" width="80%" style="background-color:#fff;" >



<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary of the Ebers–Moll model**

* The **Ebers–Moll** model fully describes the behavior of a BJT in all operating regions.
* For ordinary analysis in the forward active region, we use simplified expressions that focus mainly on the emitter–base junction.
* The **Early effect** describes the change in the effective base width of a BJT as the voltage between base and collector $V_{CB}$ changes, with $V_{BE}=\text{const}$.
* The **Early effect** improves the accuracy of current calculations in real transistor circuits.

# **II.D — Small-Signal Modeling of the BJT**


Small-signal modeling is essential for understanding and analyzing how a BJT operates **as an amplifier**, when the applied signals are **much smaller than the DC bias voltages**.


## **Purpose**

* Obtain a **linear description** of the transistor around the operating point, that is, the Q-point.
* Calculate **voltage gain**, **input/output impedance**, and **frequency response**.
* The models are based on **differentiating the nonlinear models**.


<div style="border:5px solid #0a0; background-color:#ddd; color:#003; border-radius: 15px; padding: 10px;">


<img src="../figs/lec8/H_pi_model.svg" alt="npn cross section" width="50%" style="background-color:#fff;" >

### **Derivation of the small-signal equations in the forward active region**

Assume a common-emitter BJT amplifier biased at a Q-point (quiescent point) where $V_{BE}=V_{BE}^{\text{DC}}$ and $I_C=I_C^{\text{DC}}$. Now suppose that the base-emitter voltage is perturbed by a small voltage $v_{be}(t)\ll V_{BE}^{\text{DC}}$, a **small sinusoidal perturbation**, which causes a change in collector current $i_c(t) \ll I_C^{\text{DC}}$. Therefore:
$$
V_{BE}(t) = V_{BE}^{\text{DC}} + v_{be}(t)
$$

$$
I_C(t) = I_C^{\text{DC}} + i_c(t)
$$

From the relation mentioned above:
$$
I_C = I_S e^{\frac{V_{BE}}{V_T}}
$$

We differentiate $I_C$ with respect to $V_{BE}$ to obtain the AC, or small-signal, component:
$$
\frac{dI_C}{dV_{BE}} = \frac{d}{dV_{BE}} \left(I_S e^{\frac{V_{BE}}{V_T}}\right) = \frac{I_S}{V_T} e^{\frac{V_{BE}}{V_T}} = \frac{I_C}{V_T}
$$

Thus:
$$
i_c(t) = \frac{dI_C}{dV_{BE}} \cdot v_{be}(t) = \frac{I_C}{V_T} \cdot v_{be}(t)
$$

**We define the transconductance $g_m$**:
$$
g_m = \frac{dI_C}{dV_{BE}} = \frac{I_C}{V_T}
$$
Therefore:
$$
i_c = g_m\; v_{be}
$$

#### **Base–emitter input impedance: $r_\pi$**

$r_\pi$ is a **small-signal differential resistance**, not the DC ratio $V_{BE}/I_B$. From the definition:
$$
i_b = \frac{i_c}{\beta} = \frac{g_m v_{be}}{\beta}
\Rightarrow v_{be} = i_b \cdot \left(\frac{\beta}{g_m}\right)
$$

Therefore:
$$
r_\pi = \frac{\partial v_{be}}{\partial i_b}
      = \frac{\beta}{g_m}
      = \frac{\beta V_T}{I_C}
      = \frac{V_T}{I_B}
$$


#### **Modification due to the Early effect**

$$
I_C = I_S e^{\frac{V_{BE}}{V_T}} \left(1 + \frac{V_{CE}}{V_A} \right)
$$


**Linearization around the Q-point**

Let us write $V_{CE}$ as:

$$
V_{CE}(t) = V_{CE}^{\text{DC}} + v_{ce}(t)
$$

Expanding $I_C$ as a Taylor series around the Q-point:

$$
i_c(t) = \frac{\partial I_C}{\partial V_{BE}} v_{be}(t) + \frac{\partial I_C}{\partial V_{CE}} v_{ce}(t)
$$

Derivatives:
1. Derivative with respect to $V_{BE}$, as before:
$$
\frac{\partial I_C}{\partial V_{BE}} = \frac{I_C}{V_T} = g_m
$$
2. Derivative with respect to $V_{CE}$:
$$
\frac{\partial I_C}{\partial V_{CE}} = \frac{I_S}{V_A} e^{\frac{V_{BE}}{V_T}} = \frac{I_C}{V_A}
$$

$$
i_c = g_m v_{be} + \frac{1}{r_o} v_{ce}
\quad \text{where} \quad r_o \approx \frac{V_A}{I_C}
$$

## Common-emitter amplifier

Voltage gain is defined as
$$ A_v = \frac{V_{\rm out}}{V_{\rm in}}, $$

whereas current gain is defined as
$$ A_i = \frac{I_{\rm out}}{I_{\rm in}}. $$

Let us apply these definitions to the case of an NPN BJT in common-emitter configuration. In this configuration, the emitter is the common reference for the input and output signals, as shown in the figure.

<img src="../figs/lec8/NPN_common_emitter.svg" alt="npn cross section" width="20%" style="background-color:#fff;" > <img src="../figs/lec8/NPN_common_emitter_degeneration.svg" alt="npn cross section" width="20%" style="background-color:#fff;" >

In both schematics, we have neglected the details of biasing, that is, how the transistor is placed at a stable DC operating point, or Q-point. In the second case we have added an emitter resistor; this is called **emitter degeneration**. **This configuration reduces the gain but increases the linearity and stability of the transistor**.

$$ A_v=\frac{-g_m\, R_C}{g_m\,R_E + 1} $$

In the first case, where the emitter resistor is absent, this reduces to $A_v=-g_m\,R_C$. In the other limiting case, where $g_m\,R_E \gg 1$, which is achieved for relatively large $R_E$, the voltage gain becomes $A_v=-R_C/R_E$. The current gain in either case is:
$$ A_i=\beta$$

### Summary

* **$g_m = \dfrac{I_C}{V_T}$**
* **$r_\pi = \dfrac{\beta}{g_m} = \dfrac{\beta V_T}{I_C}$**
* **$r_o = \dfrac{V_A}{I_C}$**
* **$i_b = \dfrac{v_{be}}{r_\pi}$**
* **$i_c = g_m\, v_{be} + \dfrac{v_{ce}}{r_o}$**
* **$A_v=\frac{-g_m\, R_C}{g_m\,R_E + 1}$**
* **$A_i=\beta$**

## **1. Hybrid-$\pi$ model ($\pi$ because the circuit shape resembles the Greek letter)**

### **Components of the hybrid-$\pi$ model for an NPN transistor in common-emitter configuration:**

| Element | Definition | Physical meaning |
| ---------------------- | ----------------------------------------------------------- | --------------------------------------- |
| $r_\pi$ | Base–emitter input impedance | $r_\pi = \frac{\beta}{g_m}$ |
| $g_m v_{be}$ | Voltage-controlled current source | $g_m = \frac{I_C}{V_T}$ |
| $r_o$ | Output resistance due to the Early effect | $r_o \approx \frac{V_A}{I_C}$ |
| $C_\pi$, $C_\mu$ | BE and BC junction capacitances at high frequencies | Not always included in the basic model |

### **Fundamental equations:**
$$
g_m = \frac{I_C}{V_T}, \quad
r_\pi = \frac{\beta}{g_m}, \quad
r_o \approx \frac{V_A}{I_C}
$$


### **Typical parameter values:**

* $I_C = 1$ mA
* $\beta = 100$
* $V_T = 25$ mV
* $V_A = 100$ V

Then:
$$
g_m = \frac{1\text{ mA}}{25\text{ mV}} = 40\text{ mS}, \quad
r_\pi = \frac{100}{40\text{ mS}} = 2.5\ \text{k}\Omega, \quad
r_o = \frac{100\text{ V}}{1\text{ mA}} = 100\ \text{k}\Omega
$$

## **2. h-parameter model**

The h-parameter model represents the transistor as a **two-port network**, relating input voltage and current to output voltage and current.

### Equations for the common-emitter configuration:

<img src="../figs/lec8/BJT_h-parameters_(generalised).svg" alt="npn cross section" width="50%" style="background-color:#fff;" >

$$
\begin{aligned}
v_b &= h_{ie} i_b + h_{re} v_c \\
i_c &= h_{fe} i_b + h_{oe} v_c
\end{aligned}
$$

| Parameter | Definition |
| ----------- | ----------------------------------------------- |
| $h_{ie}$ | Input impedance |
| $h_{fe}$ | Current gain ($\approx \beta$) |
| $h_{oe}$ | Output conductance ($\approx \frac{1}{r_o}$) |
| $h_{re}$ | Reverse voltage feedback, usually very small |

This model is used less often for analytical circuit analysis, but it is useful for experimental characterization.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary**

| Model | Advantages | Where it is used |
| ------------ | -------------------------------------- | --------------------------------- |
| hybrid-π | Accurate and physically based | Circuit design, simulations |
| h-parameter | Practical for experimental characterization | Simpler manual analysis |

</div>

#### **Practical strategy for analyzing a BJT circuit**

1. Remove the AC signal and analyze only the DC circuit.
2. Find $V_B$, $V_E$, $I_E$, $I_C$, and $V_{CE}$.
3. Check the operating region: cutoff, active, or saturation.
4. If the transistor is in the active region, build the small-signal model.
5. Calculate $g_m$, $r_\pi$, and, if needed, $r_o$.
6. Find the gain, input resistance, and output resistance.
7. Check whether the signal can appear at the output without clipping or distortion.

# **II.E — Practical Applications of the BJT in Circuits**


## **Purpose of this section**

We now show **how a BJT is used in practice**, both as a **signal amplifier** and as a **switch** in electronic circuits. These two functions form the basis of BJT use in **analog** and **digital** electronics.


## **1. BJT as an amplifier**

<img src="../figs/lec8/common_EBC.jpeg" alt="npn cross section" width="50%" style="background-color:#fff;" >

The transistor is used as a **voltage amplifier** in different configurations.

### **1.1 Common-emitter amplifier**

* The most common configuration.
* Gives the **largest voltage gain** and **phase inversion** of 180°.

#### General schematic idea:

* The input is applied to the **base**.
* The output is taken from the **collector**.
* The **emitter** is at AC ground, either directly or through a bypass capacitor.

#### Voltage gain:
$$
A_v = \frac{v_{out}}{v_{in}} \approx -g_m (R_C \parallel r_o)
$$

#### Characteristics:
* High gain.
* Phase inversion.
* Input impedance $r_\pi$.
* Output impedance $r_o$.


### **1.2 Common-collector amplifier / buffer**

* Input at the base, output from the emitter.
* **Voltage gain close to 1**; therefore, it does not amplify voltage but it does amplify current.
* High input impedance and low output impedance.

#### Gain:
$$
A_v \approx \frac{r_E}{r_E + r_\pi / (\beta + 1)} \approx 1
$$

#### Use:
* As a **buffer** for isolating/decoupling circuits with different loads.


### **1.3 Common-base amplifier**

* Input at the **emitter**, output at the **collector**, with the base at AC ground.
* High voltage gain, **no phase inversion**.
* Very low input impedance.

#### Use:
* Used in **high-frequency** circuits, for example RF circuits.


## **2. BJT as a switch**

When used as a **switch**, the BJT operates in **two regions**:

* **Cutoff** → the transistor is **off**.
* **Saturation** → the transistor is **on**.

### Typical application:

#### LED control circuit:

* The **base** is connected through a resistor to a control signal.
* When the signal is **high**: the BJT enters **saturation** → the LED turns on.
* When the signal is **low**: the BJT is in **cutoff** → the LED turns off.


#### Practical calculation:

1. Choose the **base resistor** to provide the required base current $I_B$:
   $$ R_B = \frac{V_{in} - V_{BE}}{I_B}, \quad \text{with} \quad I_B = \frac{I_C}{\beta_{forced}} $$
2. Here $\beta_{forced}$ is usually chosen smaller than the nominal value, for safety.


## **Common practical examples**

| Application | Role of the BJT | Region |
| ---------------------- | ------------------- | ---------------- |
| Audio amplifier circuit | Voltage amplifier | Forward active |
| LED driver | ON/OFF switch | Saturation / Cutoff |
| TTL logic circuit | Logic gate | Switching |
| Active sensor circuit | Amplifies the signal | Forward active |


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Summary**

* A BJT can be used for **amplification** in the forward active region and for **switching** in cutoff/saturation.
* The **common-emitter** configuration is the strongest for voltage amplification.
* **Gain** and **impedance** depend on the configuration and biasing.
* When used as a switch, correct dimensioning of the base current is essential for reliable control.

# **II.F — Practical Reading of BJT Datasheet Parameters**

In introductory analysis we use simple models, but in real design we must check the component datasheet. The most important parameters are:

| Parameter | Practical meaning |
|---|---|
| $V_{CEO}$ | maximum collector–emitter voltage when the base is open |
| $I_C$ max | maximum collector current |
| $P_D$ | maximum power the component can dissipate |
| $h_{FE}$ or $\beta$ | DC current gain; it has a wide spread from component to component |
| $V_{CE(sat)}$ | collector–emitter voltage when the BJT is used as a closed switch |
| $f_T$ | transition frequency, important for fast signals |
| $R_{\theta JA}$ | package-to-ambient thermal resistance |

<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

#### **Practical warning**

Do not design a circuit by treating $\beta$ as a fixed value. A datasheet usually gives a wide interval for $h_{FE}$. For this reason, good biasing circuits should depend as little as possible on the exact value of $\beta$.

</div>


# **Summary of Lecture 10**

In this lecture, we built the conceptual chain:

$$
\text{physical structure} \rightarrow \text{operating region} \rightarrow \text{model} \rightarrow \text{circuit}
$$

Main points:

- The BJT is a bipolar transistor because it involves both electrons and holes.
- In the forward active region, the BE junction is forward-biased while the BC junction is reverse-biased.
- For amplification, we usually use the forward active region.
- For switching, we use cutoff and saturation.
- Practical analysis begins with the DC operating point and then moves to the small-signal model.
- The most important small-signal parameters are $g_m$, $r_\pi$, and $r_o$.


# **Exercises**

### **1. Common-emitter amplifier with voltage-divider bias**

**Problem.** An NPN BJT is given in a common-emitter configuration. The parameters are:

* $V_{CC} = 12\,\text{V}$
* $R_1 = 100\,\text{k}\Omega$
* $R_2 = 18\,\text{k}\Omega$
* $R_C = 4.7\,\text{k}\Omega$
* $R_E = 1\,\text{k}\Omega$
* $\beta = 100$
* $V_{BE} \approx 0.7\,\text{V}$

Find:

a) the operating point $I_C$ and $V_{CE}$;  
b) the parameters $g_m$ and $r_\pi$;  
c) an approximate voltage gain if the emitter is bypassed for the AC signal.

---

**Solution.**

First, find the base voltage from the voltage divider:

$$
V_B = V_{CC}\frac{R_2}{R_1+R_2}
= 12\frac{18}{100+18}
\approx 1.83\,\text{V}
$$

The emitter voltage is:

$$
V_E = V_B - V_{BE} \approx 1.83 - 0.7 = 1.13\,\text{V}
$$

Therefore:

$$
I_E = \frac{V_E}{R_E} \approx \frac{1.13}{1\,\text{k}\Omega}
= 1.13\,\text{mA}
$$

For large $\beta$:

$$
I_C \approx I_E \approx 1.13\,\text{mA}
$$

The collector voltage is:

$$
V_C = V_{CC} - I_C R_C
\approx 12 - (1.13\,\text{mA})(4.7\,\text{k}\Omega)
\approx 6.69\,\text{V}
$$

Therefore:

$$
V_{CE} = V_C - V_E \approx 6.69 - 1.13 = 5.56\,\text{V}
$$

This is a reasonable Q-point for amplification because the transistor is far from cutoff and far from saturation.

The small-signal parameters are:

$$
g_m = \frac{I_C}{V_T}
\approx \frac{1.13\,\text{mA}}{25.9\,\text{mV}}
\approx 43.6\,\text{mS}
$$

$$
r_\pi = \frac{\beta}{g_m}
\approx \frac{100}{43.6\,\text{mS}}
\approx 2.3\,\text{k}\Omega
$$

If the emitter is bypassed for the AC signal and we neglect $r_o$, then:

$$
A_v \approx -g_m R_C
\approx -(43.6\,\text{mS})(4.7\,\text{k}\Omega)
\approx -205
$$

In practice, this gain is reduced by the load, source resistance, $r_o$, and by the requirement that the signal must not drive the transistor into cutoff or saturation.

### **2. BJT as a switch for an LED**

**Problem.** A microcontroller with a $5\,\text{V}$ output drives the base of an NPN BJT in order to turn on an LED. Assume:

* $V_{CC} = 5\,\text{V}$
* $I_C \approx 10\,\text{mA}$
* $V_{BE} \approx 0.7\,\text{V}$
* in saturation, we want to use $\beta_{forced} \approx 10$

Find an approximate value for the base resistor $R_B$.

**Solution.**

For reliable saturation, we do not use the nominal $\beta$. We use a smaller forced value:

$$
I_B \approx \frac{I_C}{\beta_{forced}}
= \frac{10\,\text{mA}}{10}
= 1\,\text{mA}
$$

Therefore:

$$
R_B \approx \frac{V_{in}-V_{BE}}{I_B}
= \frac{5-0.7}{1\,\text{mA}}
\approx 4.3\,\text{k}\Omega
$$

A practical standard value is:

$$
R_B \approx 4.7\,\text{k}\Omega
$$
