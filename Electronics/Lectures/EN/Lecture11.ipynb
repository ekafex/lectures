## **Main idea of the lecture**

In Lecture 10 we saw that the BJT controls current through carrier injection and base current.  
In this lecture we move to **field-effect transistors**, where current is controlled mainly by the **electric field** created by the gate voltage.

Ideally:

$$
I_G \approx 0
$$

therefore FETs/MOSFETs behave as devices with **very high input impedance**.

</div>



## **Orientation summary: JFET and MOSFET**

### **JFET**

A JFET has a conducting channel that is already present in its natural state. The gate controls the effective width of the channel through the reverse bias of a p–n junction.

For an n-channel JFET:

* $V_{GS}=0$ → the channel is maximally open;
* $V_{GS}<0$ → the depletion region expands and the channel narrows;
* $V_{GS}\leq V_{GS(off)}$ → the channel closes and $I_D \approx 0$.

In the active/pinch-off region:

$$
I_D = I_{DSS}\left(1-\frac{V_{GS}}{V_{GS(off)}}\right)^2.
$$

### **MOSFET**

A MOSFET has an oxide-insulated gate. The voltage $V_{GS}$ creates an electric field that forms or modifies the channel.

For an n-channel enhancement MOSFET:

| Region | Conditions | Behavior |
|---|---|---|
| Cutoff | $V_{GS}<V_{th}$ | no channel, $I_D \approx 0$ |
| Linear/Triode | $V_{GS}>V_{th}$, $V_{DS}<V_{GS}-V_{th}$ | resistive behavior |
| Saturation/Active | $V_{GS}>V_{th}$, $V_{DS}\geq V_{GS}-V_{th}$ | current source controlled by $V_{GS}$ |

<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

#### **Terminology warning**

For a **MOSFET**, “saturation” is usually the amplification region.  
For a **BJT**, “saturation” is usually the closed-switch region.

These two uses of the word “saturation” do not have the same physical meaning.

# **III.A — Field-Effect Transistor (FET): Types and Physical Structure**

## **Purpose of this section**

We introduce the **FET (Field-Effect Transistor)** as a device that, unlike the BJT, controls current through an **electric field** instead of carrier injection.

The FET is fundamental to many modern systems, especially the **MOSFET**, which makes up most transistors in digital electronics, for example in microprocessors.

## **1. Fundamental principle of the FET**

* A FET is a **three-terminal device**:
  * **Gate (G)**: the terminal that controls the electric field
  * **Source (S)**: the source of carriers
  * **Drain (D)**: the destination of carriers
* Carriers flow through a **conducting channel**, which is controlled by the voltage at the **Gate (G)**.
* Ideally, no gate current is required, only a **voltage** → **voltage-controlled device**.

## **2. Main types of FET**

### **JFET (Junction FET)**

* The channel is always present (depletion-mode behavior).
* A negative voltage at the gate narrows the channel in an n-JFET.
* The structure is based on a **p–n junction** that controls conductivity.
* The channel current is **controlled by reverse-biasing a p–n junction**, which creates a **depletion region**.

#### Main types:

* **n-channel JFET** – the channel is **n-type**, the gate is **p-type**
* **p-channel JFET** – the channel is **p-type**, the gate is **n-type**


We will focus on the **n-channel JFET**, because it is more commonly used and more intuitive. The p-channel JFET is analogous, but with reversed voltage polarities and current direction.


* **Gate (G)**: p-type region
* **Drain (D)** and **Source (S)**: terminals of the n-type channel
* The **channel** is permanent — its conductivity is modified by the electric fields created by gate biasing.

<img src="../figs/lec8/jfet_on_off.png" alt="JFET on/off" width="80%" style="background-color:#fff;" >


#### **JFET operating regions**

* When $V_{GS} = 0$, the channel is **fully open and conducting** → maximum current $I_{DSS}$
* When $V_{GS} < 0$, the p–n junction is **reverse-biased** → the depletion region expands and **narrows the channel**
* At a critical value $V_{GS} = V_P$ (pinch-off voltage), the **channel closes** → $I_D \approx 0$


| Region | Conditions on $V_{GS}$ and $V_{DS}$ | Physical behavior |
| ---------------------------------------- | ----------------------------------------------------------------- | ------------------------------------------------------------------------------- |
| **Active region (saturation/pinch-off)** | $V_{GS} < 0$ and $V_{DS} > V_{GS} - V_P$ | The channel narrows near the drain → $I_D$ is almost constant |
| **Linear region (ohmic)** | $V_{DS}$ is small, usually $V_{DS} < V_{GS} - V_P$ | The channel conducts → behaves like a controlled resistance |
| **Cutoff region** | $V_{GS} \leq V_P$ | The channel is fully closed → $I_D \approx 0$ |



<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

#### **Sign warning for n-channel JFETs**

For an n-channel JFET, usually $V_P = V_{GS(off)} < 0$.  
Therefore the formula

$$
I_D = I_{DSS}\left(1-\frac{V_{GS}}{V_P}\right)^2
$$

must be used carefully. In self-biasing, usually $V_G = 0$ and $V_S = I_D R_S$, therefore

$$
V_{GS} = -I_D R_S.
$$

In this case the equation becomes

$$
I_D = I_{DSS}\left(1-\frac{I_D R_S}{|V_P|}\right)^2.
$$

</div>


<img src="../figs/lec8/JFET_n-channel_en.svg" alt="n-channel JFET" width="90%" style="background-color:#fff;" >


#### **Cutoff Region**
* **Condition:**
  $$
  V_{GS} \leq V_P
  $$
  where $V_P$ is the **pinch-off voltage**, usually negative for an n-channel JFET.
* **What happens:**
  The depletion region expands so much that it **fully closes the channel** → **no current** flows from drain to source.

#### **Active Region (Pinch-Off / Saturation)**
* **Condition:**
  $$
  V_P < V_{GS} < 0 \quad \text{and} \quad V_{DS} \geq V_{GS} - V_P
  $$
* **What happens:**
  * Increasing $V_{DS}$ no longer significantly affects $I_D$.
  * The channel is “pinched off” near the drain → **the current saturates**, but not in the BJT sense.
* **The current is described by:**
  $$
  I_D = I_{DSS} \left(1 - \frac{V_{GS}}{V_P} \right)^2
  $$


#### **Linear Region (Ohmic Region)**
* **Condition:**
  $$
  0 < V_{DS} < V_{GS} - V_P
  $$
* **What happens:**
  * The channel is **open** and the current is **approximately linear with $V_{DS}$**.
  * The behavior is like a **resistance controlled by $V_{GS}$**.
* **Current:**
  $$
  I_D \approx k \left[ (V_{GS} - V_P) V_{DS} - \frac{V_{DS}^2}{2} \right]
  $$


### **Advantages and disadvantages of JFETs**

Advantages:
* **High input impedance**
* **Very low noise**
* **Simple design**
* Stable at high temperatures

Disadvantages:
* Not easily adapted to modern integrated technologies (integrated circuits)
* Limited flexibility for logic-circuit design
* $I_D$ control is not very linear.

### **Typical applications**
* Small-signal amplifiers at high frequencies
* Input stages for low-noise amplifiers
* Constant-current sources
* Analog switches and controlled resistances.


<br>
<br>





### **MOSFET (Metal-Oxide-Semiconductor FET)**

* It has an **insulating oxide** layer between the gate and the semiconductor body.
* The gate has **no direct contact** with the channel → **high input impedance**.

#### Two main modes:

| MOSFET type | Behavior with no gate voltage |
| -------------------- | ------------------------------------------------------------------------- |
| **Enhancement-mode** | The channel **does not exist** without $V_{GS}$ → an electric field is needed to create it |
| **Depletion-mode** | The channel exists and can be **closed** with a negative $V_{GS}$ |


<img src="../figs/lec8/FET_Symbols.svg" alt="npn cross section" width="90%" style="background-color:#fff;" >


## **3. MOSFET structure (n-channel, enhancement-mode)**


### When $V_{GS} > V_{th}$:

* Electrons are attracted to the surface → an **n-type channel** is created.
* The current $I_D$ flows from **Drain** to **Source**.


## **4. Comparative summary: JFET vs MOSFET**

| Feature | JFET | MOSFET |
| --------------------- | ------------------- | -------------------------- |
| Gate insulation | p–n junction | Oxide insulation (SiO₂) |
| Input impedance | Medium to high | Very high (10⁶–10¹² Ω) |
| Operating mode | Depletion | Enhancement or Depletion |
| Mass production | Rarely used today | Universal use in ICs |

IC = Integrated Circuits!

####  **Summary**

* A FET controls current flow through **an electric field**, not through current as in the BJT.
* The most widely used type is the **MOSFET**, because of:
  * high input impedance
  * low power consumption
  * scalability for integrated circuits

# **III.B — MOSFET Operating Regions**

## **Purpose of this section**

We explain **how the behavior of a MOSFET changes depending on the voltages $V_{GS}$ (Gate–Source)** and **$V_{DS}$ (Drain–Source)**. These regions determine **whether the channel is closed, partially formed, or fully formed**, and what characteristics the current $I_D$ has.

We will focus on the **n-channel enhancement-mode MOSFET**, because it is the most widely used in practice.


## **1. Relevant voltages**

* $V_{GS}$ — controls channel formation.
* $V_{th}$ — threshold voltage, typically 1–2 V.
* $V_{DS}$ — applied voltage that drives current flow from drain to source.



<div style="border:3px solid #b55; background-color:#fff3f0; color:#300; border-radius: 12px; padding: 12px;">

#### **Practical warning: $V_{th}$ does not mean “MOSFET fully ON”**

$V_{th}$ is usually measured at a very small drain current.  
In practical design, especially for switches driven by microcontrollers, the condition $V_{GS} > V_{th}$ is not enough.

For switching, the datasheet must be checked for:

$$
R_{DS(on)}
$$

at the real gate-drive voltage, for example $V_{GS}=3.3\,\text{V}$, $5\,\text{V}$, or $10\,\text{V}$.

</div>


## **2. Three main operating regions**


### **1. Cutoff Region**

**Condition:**
$$ V_{GS} < V_{th} $$

**What happens:**
* No conducting channel is formed.
* The current $I_D \approx 0$.
* The transistor is **OFF**.

**Use:** as an **OFF switch** in a digital circuit.


### **2. Linear Region (Triode / Linear Region)**

**Condition:**
$$ V_{GS} > V_{th} \quad \text{and} \quad V_{DS} < V_{GS} - V_{th} $$

**What happens:**
* A **fully conducting** channel is formed.
* The current depends on $V_{DS}$ and is **approximately linearly controllable**.
* The behavior is like a **voltage-controlled resistance**.

**Current:**
$$ I_D = \mu_n\, C_{ox}\, \frac{W}{L}\, \left[ (V_{GS} - V_{th}) V_{DS} - \frac{V_{DS}^2}{2} \right] $$

**Use:** as a **variable resistance** or in low-frequency amplifier structures.

### **3. Saturation Region (Active Region)**

**Condition:**
$$ V_{GS} > V_{th} \quad \text{and} \quad V_{DS} \geq V_{GS} - V_{th} $$

**What happens:**
* The channel “narrows” near the drain → **pinch-off** occurs.
* The current becomes **approximately independent of $V_{DS}$**.
* The transistor acts as a **voltage-controlled current source**.

**Current:**
$$ I_D = \frac{1}{2} \mu_n\, C_{ox}\, \frac{W}{L}\, (V_{GS} - V_{th})^2 $$

**Use:** in **analog amplifier circuits**.


<img src="../figs/lec8/IvsV_mosfet.svg" alt="npn cross section" width="70%" style="background-color:#fff;" >



<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Summary of the n-channel enhancement-mode MOSFET regions**

| Region | Condition | Behavior | Use |
| --------------- | ---------------------------------- | --------------- | ---------- |
| Cutoff | $V_{GS} < V_{th}$ | Closed channel | OFF switch |
| Linear (Triode) | $V_{DS} < V_{GS} - V_{th}$ | Conducting channel | Resistance |
| Saturation | $V_{DS} \geq V_{GS} - V_{th}$ | Constant current | Amplifier |


* A MOSFET has three basic regions, determined by $V_{GS}$ and $V_{DS}$.
* The **saturation** region is the one used for **amplification**.
* The **linear (triode)** region behaves like a **voltage-controlled resistance**.
* The **cutoff** region makes the transistor non-conducting.


**Parallel with the BJT**

| BJT | MOSFET |
| ---------------------------- | ------------------------------- |
| $I_B$ controls $I_C$ | $V_{GS}$ controls $I_D$ |
| Active region | Saturation region |
| Cutoff region | Cutoff region |
| Current control | Voltage control |


</div>

# **III.C — Mathematical Modeling of the MOSFET**

## **Purpose of this section**

We formulate the **mathematical equations for the drain current $I_D$** as a function of the control voltages $V_{GS}$ and $V_{DS}$ for the three MOSFET operating regions.

As mentioned, the focus will be on the **n-channel enhancement-mode MOSFET**, because of its wide use in digital and analog electronics.


## **1. Physical parameters of the model**

| Parameter | Meaning |
| ------------ | ------------------------------------------------ |
| $\mu_n$ | Electron mobility in the n-channel |
| $C_{ox}$ | Capacitance per unit area of the oxide layer |
| $W$, $L$ | Channel width and length |
| $V_{th}$ | Threshold voltage |
| $V_{GS}$ | Gate–source voltage |
| $V_{DS}$ | Drain–source voltage |
| $I_D$ | Current from drain to source |


## **2. Equations by operating region**

### a) **Cutoff Region**

Condition:
$$ V_{GS} < V_{th} $$

Current:
$$ I_D \approx 0 $$


### b) **Linear Region (Triode)**

Condition:
$$ V_{GS} > V_{th} \quad \text{and} \quad V_{DS} < V_{GS} - V_{th} $$

Current:
$$ I_D = \mu_n C_{ox} \frac{W}{L} \left[ (V_{GS} - V_{th}) V_{DS} - \frac{V_{DS}^2}{2} \right] $$

This is the region where the MOSFET behaves as a **voltage-controlled resistance**.


### c) **Saturation Region**

Condition:
$$ V_{GS} > V_{th} \quad \text{and} \quad V_{DS} \geq V_{GS} - V_{th} $$

Current:
$$ I_D = \frac{1}{2}\, \mu_n\, C_{ox} \,\frac{W}{L}\, (V_{GS} - V_{th})^2 $$

In this region, the current is **approximately independent of $V_{DS}$** and is controlled only by $V_{GS}$. For this reason, the MOSFET is described as a **voltage-controlled current source**.


## **3. Summary model for the current $I_D$**

For all regions in one formula:
$$
I_D = 
\begin{cases}
0, & V_{GS} < V_{th} \\
\mu_n C_{ox} \frac{W}{L} \left[(V_{GS} - V_{th}) V_{DS} - \frac{V_{DS}^2}{2} \right], & V_{DS} < V_{GS} - V_{th} \\
\frac{1}{2} \mu_n C_{ox} \frac{W}{L} (V_{GS} - V_{th})^2, & V_{DS} \geq V_{GS} - V_{th}
\end{cases}
$$


## **4. Definition of design factors**

The factor below is often used:
$$
k' = \mu_n C_{ox}
\quad \text{and} \quad
k = k' \cdot \frac{W}{L}
$$

Then the formulas become more compact:

* **Linear region**:
$$ I_D = k\; \left[(V_{GS} - V_{th}) V_{DS} - \frac{V_{DS}^2}{2} \right] $$

* **Saturation region**:
$$ I_D = \frac{1}{2}\, k \,(V_{GS} - V_{th})^2 $$


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Summary**

| Region | Condition | $I_D$ |
| ------- | ---------------------------------- | -------------------------------------- |
| Cutoff | $V_{GS} < V_{th}$ | $I_D \approx 0$ |
| Linear | $V_{DS} < V_{GS} - V_{th}$ | $I_D \propto V_{DS}$ |
| Saturation | $V_{DS} \geq V_{GS} - V_{th}$ | $I_D \propto (V_{GS} - V_{th})^2$ |

</div>

# **III.D — MOSFET Small-Signal Modeling**

## **Purpose of this section**

We describe the **linear model of the MOSFET** when it is used in the saturation region to analyze **small AC signals** — analog amplifiers, filtering circuits, voltage amplifiers, or current amplifiers.

This model is analogous to the **hybrid-$\pi$** model used for the BJT.

## **1. Fundamental assumptions**

* The MOSFET is in the **saturation region**:
  $$ V_{GS} > V_{th},\quad V_{DS} > V_{GS} - V_{th} $$
* The circuit is biased at an **operating point (Q-point)**, and a **small AC signal** is applied around this point.
* The variations are sufficiently small to linearize the equation $I_D = f(V_{GS})$.

## **2. Elements of the MOSFET small-signal model**

| Symbol | Description | Formula |
| ----------- | ------------------------------------------------ | ------------------------------------------------------------------------------------ |
| $g_m$ | Transconductance, analogous to “gain” | $g_m = \frac{\partial I_D}{\partial V_{GS}} = \frac{2 I_D}{V_{GS} - V_{th}}$ |
| $r_{ds}$ | Output resistance from channel-length modulation | $r_{ds} = \frac{1}{\lambda I_D}$ |
| $v_{gs}$ | Differential gate–source voltage | — |
| $i_d$ | Small-signal channel current | $i_d = g_m v_{gs}$ |

> Here $\lambda$ is the channel-length modulation parameter, typically very small ($\lambda \approx 0.01$ V⁻¹).


## **3. Representative small-signal equivalent circuit**



* **There is no input current into the gate** → ideal input impedance: $R_{in} = \infty$
* The **current source controlled by $v_{gs}$** determines the output.
* **$r_{ds}$** is placed in parallel with the drain load.


## **4. Voltage gain of a Common Source amplifier**

For a simple amplifier with a **resistor $R_D$ at the drain**, the gain is:
$$ A_v = \frac{v_{out}}{v_{in}} = -g_m (R_D \parallel r_{ds}) $$

If $r_{ds} \gg R_D$ is assumed, then:
$$ A_v \approx -g_m \, R_D $$

* The minus sign indicates **phase inversion**.
* The gain depends only on $g_m$ and the load resistance.


## **5. Practical numerical example**

Suppose:
* $I_D = 1$ mA
* $V_{GS} - V_{th} = 1$ V
* $\lambda = 0.02$ V⁻¹

Then:
$$ g_m = \frac{2 I_D}{V_{GS} - V_{th}} = \frac{2 \cdot 1\text{ mA}}{1\text{ V}} = 2\text{ mS} $$

$$ r_{ds} = \frac{1}{\lambda I_D} = \frac{1}{0.02 \cdot 1\text{ mA}} = 50\text{ k}\Omega  $$

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Summary**

* The **MOSFET small-signal model** is essential for AC analysis in amplifiers.
* The gain is determined by $g_m$ and the load at the drain.
* The input impedance is very high → ideal for buffers.
* The channel-length modulation effect ($\lambda$) is included through $r_{ds}$, which improves accuracy.

</div>

# **III.E — Practical MOSFET Circuit Applications**


## **Purpose of this section**

We show how the **MOSFET** is used in circuits for **analog amplification** and **digital switching**, by analyzing the classical configurations:

* **Common Source (CS)** — voltage amplifier with phase inversion
* **Common Drain (CD)** — voltage buffer
* **Common Gate (CG)** — used at high frequencies


## **1. MOSFET as an amplifier**

### **1.1 Common Source Amplifier**

* **Input at the gate**, **output at the drain**, **source grounded**.
* Behaves as a voltage amplifier with **phase inversion**.
* Increasing $V_{in}$ → increases $I_D$ → increases the voltage drop across $R_D$ → decreases $V_{out}$.

#### Voltage gain:

$$ A_v = \frac{v_{out}}{v_{in}} = -g_m \; (R_D \parallel r_{ds}) $$

#### Characteristics:
* High gain.
* Very high input impedance.
* Output impedance approximately $R_D$.


### **1.2 Common Drain Amplifier (Common Drain / Buffer)**

* **Input at the gate**, **output at the source**, **drain fixed at a voltage ($V_{DD}$)**.
* The voltage gain is **close to 1**.
* It does not amplify voltage, but it amplifies **current drive capability**.

#### Gain:
$$ A_v \approx \frac{g_m R_S}{1 + g_m R_S} \approx 1 \quad \text{if } g_m R_S \gg 1
$$

#### Characteristics:
* Very high input impedance.
* Low output impedance.
* Used as a **voltage buffer**.


### **1.3 Common Gate Amplifier**

* **Input at the source**, **output at the drain**, **gate is AC-grounded**.
* No phase inversion.
* Used at **high frequencies** because of its low input capacitance effect.

#### Gain:
$$ A_v = g_m\; (R_D \parallel r_{ds}) $$

#### Characteristics:
* Low input impedance.
* Moderate output impedance.
* Higher bandwidth than the common-source stage.


## **2. MOSFET as a Switch**

### Regions used:

* **OFF**: $V_{GS} < V_{th}$ → no channel forms → no current.
* **ON**: $V_{GS} > V_{th}$ and $V_{DS} \ll V_{GS} - V_{th}$ → high conductivity in the triode region.

### Typical applications:

* **Logic inverter (NOT gate)** with a MOSFET.
* **Driving LEDs, motors, and relays**.


<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

#### **MOSFET as a switch: what do we actually check?**

For a practical switch, we do not only check whether a channel exists. We check whether the losses are acceptable:

$$
P_{cond} \approx I_D^2 R_{DS(on)}.
$$

If the switching frequency is high, the gate charge $Q_g$ and dynamic losses must also be checked.

</div>

#### Design calculations:
* Select $V_{GS}$ so that the device moves from cutoff to triode/ON.
* Estimate $R_{DS(on)}$ to minimize losses.


<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Summary of configurations**

| Configuration | Gain $A_v$ | Phase | $R_{in}$ | $R_{out}$ | Use |
| -------------------- | ---------------------------------- | ---- | ------------- | -------- | --------------------------- |
| Common Source (CS) | $-g_m (R_D \parallel r_{ds})$ | 180° | Very high | $R_D$ | Voltage amplifier |
| Common Drain (CD) | Close to 1 | 0° | Very high | Low | Buffer (impedance matching) |
| Common Gate (CG) | $g_m (R_D \parallel r_{ds})$ | 0° | Low | Moderate | High frequencies |


* **Common Source** provides voltage gain with phase inversion.
* **Source Follower** is a buffer — used for impedance matching.
* **Common Gate** provides gain without inversion and is used at high frequencies.
* As a **switch**, the MOSFET is superior because it is controlled with almost no static input power.


**Comparison with BJT**

| Aspect | BJT | MOSFET |
| ------------------- | ------------------- | -------------------------- |
| Control | By current ($I_B$) | By voltage ($V_{GS}$) |
| Input impedance | low | very high |
| Power consumption | higher | very low |
| Speed | lower in ICs | very high (CMOS, GHz+) |

</div>

# **VI — Comparison Between BJT and MOSFET**

## **Purpose of this section**

We make a complete comparison between the **bipolar transistor (BJT)** and the **field-effect transistor (MOSFET)** from the physical, functional, and practical points of view, to help choose the right device in electronic design.


## **1. Comparison of the operating principle**

| Aspect | BJT | MOSFET |
| --------------------- | ------------------------------------------- | ------------------------------------------------ |
| Main mechanism | Carrier injection and diffusion | Channel modulation by electric field |
| Control | By **current** ($I_B$ controls $I_C$) | By **voltage** ($V_{GS}$ controls $I_D$) |
| Carrier transport | **Bipolar** (electrons and holes) | **Unipolar** (majority carriers only) |
| Carrier type | Electrons + holes | Electrons (n-MOS) or holes (p-MOS) |

## **2. Electrical parameters**

| Parameter | BJT | MOSFET |
| ---------------------- | --------------------------------- | ----------------------------------- |
| Input impedance | Low (kΩ) | Very high (MΩ–TΩ) |
| Input power consumption | Present ($I_B$) | Negligible (gate current ≈ 0) |
| Current gain | $I_C = \beta I_B$ | No current gain in the gate-control sense |
| Voltage gain | High in certain configurations | High in certain configurations |
| Output resistance | Moderate ($r_o$) | High ($r_{ds}$, but variable) |
| Characteristic turn-on voltage | $V_{BE} \approx 0.6-0.7$ V | $V_{th}$ typically 1–2 V |


## **3. Dynamic and frequency behavior**

| Aspect | BJT | MOSFET |
| ------------------------- | ----------------------------- | ---------------------------------- |
| Switching speed | Slower because charge is stored | Faster; input is capacitive |
| Parasitic capacitances | Lower | Present: $C_{gs}, \; C_{gd}$ |
| Switching frequency | Up to 100 MHz | Up to GHz |


## **4. Practical use**

| Application | More suitable device |
| ----------------------------------------------------- | ------------------------- |
| Audio amplifier with high gain and low noise | **BJT** |
| Digital switching, CMOS logic | **MOSFET** |
| High-density integrated circuits (ICs) | **MOSFET** (CMOS) |
| Small-signal RF amplifiers | **BJT** (CB) / **MOSFET CG** |
| Simple discrete analog circuits with stable biasing | **BJT** |


## **5. Comparative summary table**

| Characteristic | BJT | MOSFET |
| ------------------ | ------------------------------- | ------------------------------------------- |
| Type of control | Current | Voltage |
| Carrier type | Bipolar (electrons + holes) | Unipolar (electrons or holes) |
| Input consumption | Base current is required | Almost zero |
| Input impedance | Low | Very high |
| Current gain | Large ($\beta \approx 100$) | None at the gate input |
| Voltage gain | High, but depends on biasing | High in CS, but strongly depends on $g_m$ |
| Speed | Limited by base charge | High because of capacitive control |
| Size in ICs | Larger | Small, easily integrated |
| Dominates in: | Analog and power electronics | Digital electronics, CMOS |

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

####  **Didactic conclusion**

* **BJT** is suitable for applications that require **low-noise amplification** and stable behavior for continuous signals.
* **MOSFET** dominates in switching, **digital electronics**, and **energy efficiency**.

The choice between them is not absolute — it depends on the application, supply voltages, power consumption, and physical size.

## **Practical selection criterion**

<div style="border:3px solid #777; background-color:#f7f7f7; color:#111; border-radius: 12px; padding: 12px;">

The choice is not made by an absolute rule. In practice, we ask:

1. Is very high input impedance required? → usually MOSFET/JFET.
2. Is efficient switching of loads required? → usually MOSFET.
3. Is very small-signal, low-noise amplification required? → JFET or BJT, depending on the case.
4. Is the device driven by a 3.3 V/5 V microcontroller? → **logic-level** MOSFET, checking $R_{DS(on)}$ in the datasheet.
5. Is it a simple discrete analog circuit? → BJT or JFET may be more suitable for learning and prototyping.

# **VII — FET/MOSFET Exercises**

## **Exercise 1 — Self-biased JFET common-source amplifier**

**Problem:** An n-channel JFET is given with the following parameters:

* $I_{DSS} = 10\,\text{mA}$
* $V_{GS(off)} = -4\,\text{V}$
* $V_{DD} = 15\,\text{V}$
* $R_D = 1\,\text{k}\Omega$
* $R_S = 470\,\Omega$
* $R_G = 1\,\text{M}\Omega$

Find:

a) the operating point $I_D$, $V_{GS}$, $V_{DS}$  
b) the approximate voltage gain $A_v$

### **Solution**

Since the gate draws no current:

$$
V_G \approx 0.
$$

The source voltage is:

$$
V_S = I_D R_S.
$$

Therefore:

$$
V_{GS}=V_G-V_S=-I_D R_S.
$$

We use Shockley’s equation:

$$
I_D = I_{DSS}\left(1-\frac{V_{GS}}{V_{GS(off)}}\right)^2.
$$

With $V_{GS}=-I_D R_S$ and $V_{GS(off)}=-4\,\text{V}$:

$$
I_D = 10\,\text{mA}\left(1-\frac{I_D R_S}{4\,\text{V}}\right)^2.
$$

The numerical solution gives approximately:

$$
I_D \approx 3.49\,\text{mA}.
$$

Therefore:

$$
V_S = I_D R_S \approx 1.64\,\text{V}
$$

and

$$
V_{GS}\approx -1.64\,\text{V}.
$$

The drain–source voltage is:

$$
V_{DS} = V_{DD} - I_D(R_D+R_S)
$$

$$
V_{DS} \approx 15 - 3.49\,\text{mA}(1.47\,\text{k}\Omega)
       \approx 9.88\,\text{V}.
$$

The transconductance at this operating point is:

$$
g_m = \frac{2I_{DSS}}{|V_{GS(off)}|}
\left(1-\frac{V_{GS}}{V_{GS(off)}}\right)
$$

$$
g_m \approx 2.95\,\text{mS}.
$$

If we neglect $r_o$ and there is no external load:

$$
A_v \approx -g_m R_D \approx -2.95.
$$

## **Exercise 2 — MOSFET as a switch for a DC load**

A MOSFET is used as a low-side switch for a load that draws:

$$
I_D = 1\,\text{A}.
$$

From the datasheet we have:

$$
R_{DS(on)} = 80\,\text{m}\Omega
$$

for $V_{GS}=4.5\,\text{V}$.

### **Required**

a) The power dissipated in the MOSFET during conduction.  
b) Is knowing only $V_{th}$ enough to decide whether this MOSFET is suitable?

### **Solution**

The conduction power loss is:

$$
P_{cond} = I_D^2 R_{DS(on)}.
$$

Thus:

$$
P_{cond} = (1\,\text{A})^2(0.08\,\Omega)=0.08\,\text{W}.
$$

This loss is small for many practical applications.

However, knowing only $V_{th}$ is not enough.  
$V_{th}$ indicates the beginning of channel formation for very small currents. For a practical switch, $R_{DS(on)}$ must be checked at the actual gate-drive voltage.

## **Exercise 3 — MOSFET common-source: small-signal model**

A MOSFET operates in saturation with:

* $I_D = 2\,\text{mA}$
* $V_{GS}-V_{th}=0.5\,\text{V}$
* $R_D = 4.7\,\text{k}\Omega$
* $r_o$ is neglected

Find the voltage gain.

### **Solution**

Transconductance:

$$
g_m = \frac{2I_D}{V_{GS}-V_{th}}
$$

$$
g_m = \frac{2(2\,\text{mA})}{0.5\,\text{V}}
= 8\,\text{mS}.
$$

For common-source:

$$
A_v \approx -g_m R_D.
$$

Therefore:

$$
A_v \approx -(8\,\text{mS})(4.7\,\text{k}\Omega)
\approx -37.6.
$$

The negative sign indicates phase inversion between input and output.



# **VIII — Practical FET/MOSFET Datasheet Reading**

## **Main MOSFET parameters**

| Parameter | Practical meaning |
|---|---|
| $V_{DS(max)}$ | maximum drain–source voltage |
| $I_D(max)$ | maximum drain current, usually thermally limited |
| $V_{GS(max)}$ | maximum gate–source voltage; exceeding it may damage the oxide |
| $V_{GS(th)}$ | threshold voltage for very small currents; **not the full turn-on voltage** |
| $R_{DS(on)}$ | ON-state resistance; the key parameter for switches |
| $Q_g$ | total gate charge; determines switching speed/losses |
| $C_{iss}$, $C_{oss}$, $C_{rss}$ | parasitic capacitances |
| $P_D$ | maximum dissipated power |
| $R_{\theta JA}$ | thermal resistance from junction to ambient |

<div style="border:3px solid #088; background-color:#e8ffff; color:#003; border-radius: 12px; padding: 12px;">

#### **Practical rule**

For a switch driven by a 3.3 V or 5 V microcontroller, choose a MOSFET whose datasheet clearly gives $R_{DS(on)}$ for that value of $V_{GS}$.

If the datasheet gives $R_{DS(on)}$ only for $V_{GS}=10\,\text{V}$, the device may not turn on properly from a microcontroller.

# **Lecture 11 Summary**

This lecture covered field-effect transistors.

Main points:

1. A **FET** controls current through an electric field, not through input current as in a BJT.
2. A **JFET** has an existing channel and is controlled by expanding/narrowing the depletion region.
3. A **MOSFET** has an insulated gate; the electric field creates or modifies the channel.
4. For an n-channel enhancement MOSFET, the main regions are:
   * cutoff,
   * linear/triode,
   * saturation/active.
5. The saturation region is usually used for amplification.
6. Cutoff and triode/ON are used for switching.
7. For practical analysis:
   * first find the DC operating point,
   * check the operating region,
   * build the small-signal model,
   * calculate the gain and check the validity of the model.
8. In practice, especially for MOSFET switches, the datasheet is as important as the ideal model.

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 15px; padding: 10px;">

#### **Main message**

The MOSFET is a voltage-controlled device, but practical design is not based on $V_{th}$ alone.  
For amplifiers, we check $g_m$, $r_o$, and the operating point.  
For switches, we check $R_{DS(on)}$, $Q_g$, dissipated power, and thermal limits.
