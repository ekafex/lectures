# **Semiconductor Physics**

## **1. Atomic and Crystal Structure of Semiconductors**

### 1.1 Atomic structure – Semiconductors as Group IV elements
Classical semiconductors such as **silicon (Si)** and **germanium (Ge)** belong to **Group IV of the periodic table**. This means that they have **four electrons in the valence shell**.

From a quantum-mechanical point of view:
- Electrons surround the nucleus in **allowed energy levels**, described by the solutions of the Schrödinger equation for the electrostatic electron–nucleus potential.
- In an isolated atom, these levels are **discrete** (the energies are quantized).

#### For example, for Silicon:
- The electronic configuration is:  
  $$
  \text{Si: } 1s^2 2s^2 2p^6 3s^2 3p^2
  $$
- The four electrons in the $n=3$ level are the ones that participate in chemical bonding — the **valence electrons**.

> These electrons are also responsible for the **electrical behavior** of the material.



### 1.2 Covalent bonding and the crystal lattice

In the solid state, silicon atoms are arranged in a **crystalline structure** similar to the **diamond crystal structure**.

#### Crystal structure:
- Each atom is bonded to **four nearest neighbors** through **covalent bonds**.
- Atoms share electron pairs with one another, forming strong bonds and a **very stable crystal lattice**.
- In this ideal state, electrons are **localized in bonds** and are not free to contribute to conductivity.

> This explains why a pure silicon crystal at very low temperature behaves as an **insulator**.

The unit cell of this crystal (the fundamental repeating unit in space) has **face-centered cubic (FCC)** symmetry, as shown in the figure.

<img src="../figs/lec6/Si_crystal.svg" alt="Si" width="400" style="background-color:#fff;" >



### 1.3 From discrete levels to energy bands

In a single atom, electrons are allowed only in a few specific energy levels.

On the other hand, in a **crystal with many nearby atoms**, the valence atomic orbitals **overlap** and interact.

#### Splitting of levels:
- When $N \sim 10^{23}$ identical atoms interact, the discrete energy levels split into $N$ very closely spaced levels.
- These levels form a **continuous energy band**.

The result:
- A **valence band** is formed (filled with electrons)
- And a **conduction band** is formed (usually empty)

> Electrons can contribute to electric current only if they are located in the **conduction band**.


<img src="../figs/lec6/Si_bands.svg" alt="Si" width="700"  style="background-color:#fff;" >


### 1.4 Periodic potential energy in the crystal

The energy felt by an electron in a crystal is **periodic**:
$$ V(x) = V(x + a), $$
where $a$ is the spatial period (the distance between lattice sites).

#### Quantum solution:
- The Schrödinger equation in this potential gives **Bloch-type solutions**:
$$ \psi_k(x) = u_k(x) e^{ikx} $$

- This structure creates:
  - **allowed energy bands**
  - **forbidden regions (band gaps)**


#### **Valence band and conduction band**

After the formation of bands, we have two important bands:

#### Valence band:
- The band resulting from the interaction of the **valence orbitals**.
- At low temperature, this band is **completely filled** with electrons.

#### Conduction band:
- The higher-energy band.
- In semiconductors, this band is **empty at $T=0 K$**, but it can be populated by electrons that gain energy (for example when $T>0$).

#### The gap between the two bands — **band gap $E_g$**:
$$ E_g = E_c - E_v $$


<img src="../figs/lec6/material_bands.svg" alt="Si" width="700"  style="background-color:#fff;" >


### Basic principles of conductivity

**A completely filled band cannot conduct current.**
- Electrons need **nearby available states in energy** in order to move.
- This happens only in:
  - partially filled bands  
  - or in the conduction band


### **Concrete examples of band gap values**

| Material       | Type          | Band Gap (eV) |
|----------------|----------------|----------------|
| Germanium (Ge) | Semiconductor  | 0.66           |
| Silicon (Si)   | Semiconductor  | 1.12           |
| Gallium Arsenide (GaAs) | Semiconductor | 1.42 |
| Carbon (diamond) | Insulator     | ∼5.5           |
| Copper (Cu)     | Conductor      | 0 (bands overlap) |

#### Observation:
- The smaller the band gap, the **more easily electrons can move into the conduction band**.

At room temperature:
$$
k_B T \approx 0.025 \text{ eV}
$$

> This energy is sufficient for a small fraction of the electrons to be excited into the conduction band.



### Connection with conductivity

#### Semiconductors:

When an electron moves into the conduction band:
- it becomes a **negative carrier (free electron)**

But in the valence band a missing electron remains:

This missing electron is called a **hole**
- The hole behaves like a **positive carrier**
- The motion of electrons in covalent bonds can be interpreted as the **motion of holes in the opposite direction**

> Therefore, a pair is created:
> - electron (in the conduction band)
> - hole (in the valence band)


#### Conduction mechanism

> Electric current in semiconductors is the result of:
> - the motion of **electrons in the conduction band**
> - and the motion of **holes in the valence band**


---


In short, the electrical behavior of a material depends on:
- the **size of the band gap**
- and on whether electrons can move from the valence band into the conduction band.

These characteristics classify materials into:
- **Conductors (Metals)**: no band gap → electrons move freely.
  * Electrons are very free → they move easily in an electric field.
  * Very high conductivity (∼10⁶ – 10⁸ S/m).
- **Insulators**: very large band gap → no motion (no free electrons)
   * Very large band gap → no free carriers.
   * Very low conductivity.
- **Semiconductors**: small band gap → some electrons can be thermally excited and move into the conduction band, where they are free to move, together with the role of holes.
  * Small band gap → some thermal carriers are created.
  * Conductivity is low, but **controllable**.

> This creates **charge carriers** that can transport current.

### Summary:

- The atomic and crystal structure determines the **energy band structure**.
- The bands determine whether **mobile electrons** can exist.
- The band gap controls the **number of charge carriers**
- The carriers in semiconductors are:
  - electrons  
  - holes  

Next:
> We will see how we can **control the number and type of these carriers** (through doping), and how structures are formed in which conductivity changes in space — such as the **p–n junction**.

## **2. Charge carriers and doping in semiconductors**

### **2.1 Conductivity in pure (intrinsic) semiconductors**

In a pure semiconductor, such as **pure silicon**, at low temperature (0 K) the conduction band is empty, while the valence band is filled.

However, as the **temperature increases**, some electrons gain **thermal energy** $(k_B T \sim 0.025 \text{ eV}$ at 300 K) and **move into the conduction band**.

This leads to:
1. A **free electron** in the conduction band  
2. A **hole** in the valence band  

$$ \text{e}^- \quad \leftrightarrow \quad \text{hole}^+ $$

> These are the **charge carriers** that can contribute to current.

#### Mechanism of motion:
- **Drift**: motion under the influence of an electric field  
- **Diffusion**: motion from a region of high concentration toward one of lower concentration  

> These two mechanisms are fundamental for the behavior of semiconductor devices.



### **2.2 Doping – controlled impurity addition to increase conductivity**

To increase conductivity, we add **controlled impurities** (~1 in 10⁷ atoms).

The material becomes **extrinsic** and the number of carriers increases significantly.

<img src="../figs/lec6/Si_doping.svg" width="900" style="background-color:#fff;" > 


### **2.3 n-type doping**

- Group V-A elements (P, As)  
- They provide **one free electron**

$$ \text{P}^0 \rightarrow \text{P}^+ + e^- $$

- Electrons are the **majority carriers**

> **The Fermi level shifts toward the conduction band**



### **2.4 p-type doping**

- Group III-A elements (B, Al)  
- They create **holes**

$$ \text{B}^0 + e^- \rightarrow \text{B}^- + \text{hole}^+ $$

- Holes are the **majority carriers**

> **The Fermi level shifts toward the valence band**



### **2.5 Total conductivity**

$$ \sigma = q (n \mu_n + p \mu_p) $$

- $n, p$ → concentrations  
- $\mu_n, \mu_p$ → mobilities  

> Conductivity depends both on the **number of carriers** and on their **ability to move**.


### **2.6 Toward p–n junctions**

Now we have two types of materials:

- **n-type** regions (many electrons)
- **p-type** regions (many holes)

When these two regions are brought into contact:
- electrons diffuse from n → p  
- holes diffuse from p → n  

This creates:
- charge separation  
- an internal electric field  

> This is the first step toward the formation of a **p–n junction**, which we will analyze next.


<img src="../figs/lec6/Pn-junction-equilibrium.svg" width="900" style="background-color:#fff;" > 


### **2.7 Formation of the depletion region**

When carriers diffuse:

- electrons that cross into the p region **recombine with holes**  
- holes that cross into the n region do the same  

In the region near the boundary:
- no free carriers remain  
- only fixed ions remain (the ionized dopants)


This creates:
- a **carrier-depleted region**
- a **charge separation**


### **2.8 Formation of the internal electric field**

The fixed charges create an **internal electric field**:
- this field opposes diffusion  
- the system reaches **equilibrium**

> This is called the **built-in potential**




### Summary

- Doping controls the **type and number of carriers**
- Transport occurs through:
  - drift  
  - diffusion
- When these regions are joined:
  - diffusion occurs  
  - an electric field is created  
  - a **p–n junction** is formed


In the following lecture:
> We will see how this structure leads to a **strongly nonlinear current behavior** and how it is described mathematically (the diode).

## **3. The p–n junction**

### **3.1 What is a p–n junction?**

When we place a **p-type** semiconductor and an **n-type** semiconductor next to one another, a **p–n junction** is formed.

> At this interface, mainly **electrostatic and carrier-statistical phenomena** take place, and these determine the behavior of the device.



### **3.2 What happens at the p–n contact?**

<img src="../figs/lec6/diode.svg" alt="Si" width="500" style="background-color:#fff;" >

#### a) Carrier diffusion:
- Electrons (from the n region) diffuse toward the p region  
- Holes (from the p region) diffuse toward the n region  

> the cause: a **concentration gradient**



#### b) Recombination:
- Electrons and holes recombine → they disappear as free carriers  
- A region with no carriers is formed → the **depletion region**



#### c) Formation of the electric field:
- Fixed ions remain:
  - negative on the p side  
  - positive on the n side  

an **internal electric field** is created
- this field opposes diffusion  
- the system reaches equilibrium



### **3.3 Potential barrier (built-in potential)**

The electric field creates an **energy barrier** that opposes the carriers:

$$
V_0 \approx \frac{kT}{q} \ln\left( \frac{N_a N_d}{n_i^2} \right)
$$

- It depends on doping and temperature  
- For silicon: ~0.6–0.7 V  

> This is why carriers do not move freely at equilibrium.



## **4. Biasing the p–n junction**

### **4.1 Forward bias**
- The external field **reduces the barrier**
- The depletion region **narrows**

Carriers can cross the boundary more easily:
- electrons from n → p  
- holes from p → n  

> Current rises very rapidly with voltage



### **4.2 Reverse bias**
- The external field **increases the barrier**
- The depletion region **widens**

> Majority carriers are blocked  
> Only a very small current flows (minority carriers)



### **4.3 Breakdown**

At high reverse voltage:
- carriers are strongly accelerated  
- they create new carriers (avalanche)  
- or tunnel through the barrier (Zener)

> The current rises very rapidly



### Summary

- Diffusion creates a carrier-free region  
- This creates an electric field and a barrier  
- Bias changes this barrier  

This structure leads to a strongly nonlinear current behavior,  
which we will describe mathematically in the following lecture (the diode).


```python
import numpy as np
import matplotlib.pyplot as plt

V = np.linspace(-1, 1, 400)
q = 1.6e-19  # C
k = 1.38e-23  # J/K
T = 300  # K
n = 1.5
Is = 1e-12

I = Is * (np.exp((q * V) / (n * k * T)) - 1)

plt.figure(figsize=(8, 4))
plt.plot(V, I * 1e6)  # mikroampera
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.title("V-I characteristics of semoconductor diode")
plt.xlabel("Voltage V (V)")
plt.ylabel("Current I (μA)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

Vin = np.linspace(-1, 1.5, 500)
R = 1000  # Ohm
Vd_th = 0.7
I_out = np.where(Vin > Vd_th, (Vin - Vd_th) / R, 0)

plt.figure(figsize=(8, 4))
plt.plot(Vin, I_out * 1e3)  # mA
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.title("Current in a circuit with a diode and a resitor")
plt.xlabel("Driving Voltage Vin (V)")
plt.ylabel("Current I (mA)")
plt.grid(True)
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display

@widgets.interact(R=widgets.FloatSlider(value=1000, min=100, max=5000, step=100, description='Resistor (Ω)'),
                  Vd_th=widgets.FloatSlider(value=0.7, min=0.2, max=1.0, step=0.05, description='Diode Threshold (V)'))
def interactive_diode(R, Vd_th):
    Vin = np.linspace(-1, 1.5, 500)
    I_out = np.where(Vin > Vd_th, (Vin - Vd_th) / R, 0)
    plt.figure(figsize=(8, 4))
    plt.plot(Vin, I_out * 1e3)
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.title("Current vs Vin for different parameters")
    plt.xlabel("Voltage in Vin (V)")
    plt.ylabel("Current I (mA)")
    plt.grid(True)
    plt.show()
```
