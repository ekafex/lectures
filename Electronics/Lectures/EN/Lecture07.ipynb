# Lecture 7 — Frequency Response, Transfer Functions, and Filter Applications

## 1. Motivation

In the previous lecture we studied **transient response**: what happens just after switching or powering a circuit that contains $C$ and/or $L$.  
Here we focus on **steady-state sinusoidal response**: what the circuit does when the input is AC with a sinusoid of angular frequency $\omega$.

This viewpoint is powerful because many real **signals can be decomposed into sinusoids of different frequencies via Fourier transform.**
Therefore, by understanding the response to a sinusoid, we understand how the circuit treats the frequency content of a signal.

The main idea is simple:
- a resistor does **not** care about frequency,
- a capacitor increasingly conducts as frequency increases,
- an inductor increasingly opposes current as frequency increases.

Because of that, circuits made from $R$, $L$, and $C$ can **select**, **attenuate**, **delay**, or **shape** signals.

## 2. Sinusoidal signals

We take the input in the form

$$
v(t)=V_0 \, \cos(\omega t+\phi).
$$

Where, $V_0$ is the amplitude and $\omega t+\phi$ is the phase.
For linear circuits, the steady-state output is also sinusoidal at the **same frequency**, but in general with:
- a different amplitude,
- a phase shift.

So the expected output has the form

$$
v_{\text{out}}(t)=V'_0 \,\cos(\omega t+\phi+\theta).
$$

Here, $V'_0$ and $\omega t+\phi+\theta$ are the output signal amplitude and phase respectively. $\theta$ is the phase difference between input and output signals. This motivates two key questions:
1. How much is the amplitude changed?
2. How much is the phase shifted?

The answer to both is contained in the **transfer function**.

## 3. Phasor representation

A sinusoid (cosine) can be represented as the real part of a complex exponential:

$$
v(t)=\Re\left\{\tilde{V} e^{j\omega t}\right\}.
$$

The complex quantity $\tilde V$ is called the **phasor**. 
It stores **amplitude and phase in a compact form**.

For example,
$$
V_0 \, \cos(\omega t+\phi)=\Re\left\{V_0\, e^{j\phi} e^{j\omega t}\right\}.
$$

So we identify the phasor as

$$
\tilde V = V_0\, e^{j\phi}.
$$

Using phasors is useful because differentiation becomes multiplication:
$$
\frac{d}{dt} \leftrightarrow j\omega.
$$

**That is the step that converts differential equations into algebraic relations in sinusoidal steady state.**

## 4. Impedance

The frequency-domain analogue of resistance is the **impedance** $Z$:

$$
\tilde V = Z \tilde I.
$$

For the basic passive elements:

### Resistor
$$
\boxed{Z_R = R}
$$

### Inductor
Starting from $v(t)=L\frac{di}{dt}$,
$$
\tilde V = j\omega L\,\tilde I
\quad \Rightarrow \quad
\boxed{Z_L=j\omega L}
$$

### Capacitor
Starting from $i(t)=C\frac{dv}{dt}$,
$$
\tilde I = j\omega C\,\tilde V
\quad \Rightarrow \quad
\boxed{Z_C=\frac{1}{j\omega C}}
$$

### Physical meaning

- **Resistor**: no energy storage, no phase shift between $v$ and $i$.
- **Inductor**: impedance grows with frequency, so it resists rapid current variation.
- **Capacitor**: impedance decreases with frequency, so it allows rapid variation more easily.

This is the origin of filtering behavior.

## 5. Transfer function

For a linear circuit, the transfer function is

$$
\boxed{H(j\omega)=\frac{\tilde V_{\text{out}}}{\tilde V_{\text{in}}}.}
$$

It is generally complex. Its two most important pieces are:

- **Magnitude**:
  $$
  |H(j\omega)|
  $$
  which tells us amplitude gain or attenuation.

- **Phase**:
  $$
  \angle H(j\omega)
  $$
  which tells us the phase shift between input and output.

If we use the Laplace variable $s$, then more generally

$$
\boxed{H(s)=\frac{V_{\text{out}}(s)}{V_{\text{in}}(s)}.}
$$

This treats both transient and steady-state solutions with $s=\sigma + j\omega$. If system operates for a long time, the transient phenomena die-out and $\sigma\to 0$. Therefore the system goes into steady state, the frequency response of which, is obtained by evaluating on the imaginary axis:

$$
s=j\omega.
$$

This connects the more general Laplace-domain viewpoint to ordinary AC analysis.

## 6. Canonical first-order filters

### 6.1 RC low-pass filter

Take a series $RC$ circuit and measure the output across the capacitor.

<img src="../figs/lec5/RC_LowPass.svg" width=400/>

Using the voltage divider with impedances:

$$
H_{LP}(j\omega)=\frac{Z_C}{R+Z_C}
=
\frac{\frac{1}{j\omega C}}{R+\frac{1}{j\omega C}}
=
\frac{1}{1+j\omega RC}.
$$

Magnitude:

$$
|H_{LP}(j\omega)|=\frac{1}{\sqrt{1+(\omega RC)^2}}
$$

Phase:

$$
\angle H_{LP}(j\omega)= -\tan^{-1}(\omega RC)
$$

Interpretation:

- for $\omega \ll 1/RC$, the output is almost the same as the input,
- for $\omega \gg 1/RC$, the output is strongly attenuated.

So the circuit passes **low frequencies** and suppresses **high frequencies**.

The cutoff angular frequency is

$$
\omega_c=\frac{1}{RC},
\qquad
f_c=\frac{1}{2\pi RC}.
$$

At cutoff, the magnitude becomes

$$
|H|=\frac{1}{\sqrt 2},
$$

which corresponds to the famous **$-3\,\text{dB}$ point**.

<br>
<br>
<br>

### 6.2 RC high-pass filter

Now take the same series $RC$ circuit but measure the output across the resistor.

<img src="../figs/lec5/RC_HighPass.svg" width=400/>


Then see the circuit (output) as a voltage divider we get:

$$
H_{HP}(j\omega)=\frac{R}{R+\frac{1}{j\omega C}}
=
\frac{j\omega RC}{1+j\omega RC}.
$$

Magnitude:

$$
|H_{HP}(j\omega)|=\frac{\omega RC}{\sqrt{1+(\omega RC)^2}}
$$

Phase:

$$
\angle H_{HP}(j\omega)=\tan^{-1}\!\left(\frac{1}{\omega RC}\right)
= \frac{\pi}{2}-\tan^{-1}(\omega RC)
$$

Interpretation:

- for $\omega \ll 1/RC$, the output is very small,
- for $\omega \gg 1/RC$, the output approaches the input.

So the circuit passes **high frequencies** and attenuates **low frequencies**.

<br>
<br>

### 6.3 RL filters

The same logic works for an $RL$ circuit.

- If the output is taken across **$R$**, the circuit behaves as a **low-pass**.
- If the output is taken across **$L$**, the circuit behaves as a **high-pass**.

<img src="../figs/lec5/RL.svg" align="center" width=400/>

For the series $RL$ circuit:

$$
Z_R = R, \qquad Z_L=j\omega L.
$$

So for output across the resistor,

$$
H_{RL,\;LP}(j\omega)=\frac{R}{R+j\omega L}.
$$

For output across the inductor,

$$
H_{RL,\;HP}(j\omega)=\frac{j\omega L}{R+j\omega L}.
$$

The cutoff frequency is now controlled by the time constant

$$
\tau = \frac{L}{R},
\qquad
\omega_c=\frac{R}{L}.
$$

## 7. Second-order behavior and band-pass response

When both $L$ and $C$ are present, the circuit can develop a preferred frequency:

$$
\omega_0=\frac{1}{\sqrt{LC}}.
$$

That is the **natural** or **resonant** angular frequency.

A second-order transfer function often has the form

$$
H(s)=\frac{K\,(\text{numerator})}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}.
$$

Two new ideas appear:

- **resonance** at or near $\omega_0$,
- **quality factor** $Q$, which tells us how sharp the response is.

A typical **band-pass** filter transmits frequencies near $\omega_0$ and suppresses frequencies well below and well above it.

Qualitatively:

- small $Q$  $\to$ broad peak,
- large $Q$  $\to$ narrow, sharp peak.

This is why RLC circuits are central in tuning, selection, and resonance phenomena.

## 8. Integrator and differentiator as limiting cases

These are not new circuits in essence; they are special operating regimes of RC filters.

### RC integrator

<img src="../figs/lec5/RC_int.svg" align="center" width=400/>

For the low-pass filter,

$$
H(j\omega)=\frac{1}{1+j\omega RC}.
$$

If $\omega RC \gg 1$, then

$$
H(j\omega)\approx \frac{1}{j\omega RC}.
$$

Thus

$$
\tilde V_{\text{out}} \approx \frac{1}{j\omega RC}\tilde V_{\text{in}}.
$$

In the time domain this corresponds to approximate integration:

$$
\boxed{v_{\text{out}}(t)\approx \frac{1}{RC}\int v_{\text{in}}(t)\,dt.}
$$

### RC differentiator

<img src="../figs/lec5/RC_dif.svg" align="center" width=400/>

For the high-pass filter,

$$
H(j\omega)=\frac{j\omega RC}{1+j\omega RC}.
$$

If $\omega RC \ll 1$, then

$$
H(j\omega)\approx j\omega RC.
$$

Hence

$$
\tilde V_{\text{out}} \approx j\omega RC\,\tilde V_{\text{in}},
$$

which in the time domain corresponds to

$$
\boxed{v_{\text{out}}(t)\approx RC\,\frac{dv_{\text{in}}}{dt}.}
$$

These approximations are useful, but only within the proper frequency range.

## 9. Bode plots and logarithmic interpretation

A **Bode plot** shows the transfer function versus frequency in two separate graphs:

1. Magnitude, often in decibels:
   $$
   G_{dB}=20\log_{10}|H(j\omega)|
   $$

2. Phase:
   $$
   \angle H(j\omega)
   $$

Why use logarithmic frequency?

- real circuits operate across many decades of frequency,
- slopes become easy to interpret,
- multiplication in magnitude becomes addition in decibels.

### Important first-order slopes

For a first-order filter:

- each pole contributes about **$-20\ \text{dB/decade}$** above its corner,
- each zero contributes about **$+20\ \text{dB/decade}$** above its corner.

This gives a rapid visual way to estimate circuit behavior even before exact calculation.

```python
import numpy as np
import matplotlib.pyplot as plt

R = 1e3
C = 1e-6
f = np.logspace(0, 6, 1000)   # 1 Hz to 1 MHz
w = 2*np.pi*f

H_lp = 1 / (1 + 1j*w*R*C)
H_hp = (1j*w*R*C) / (1 + 1j*w*R*C)

fc = 1/(2*np.pi*R*C)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.semilogx(f, 20*np.log10(np.abs(H_lp)), label='RC low-pass')
plt.semilogx(f, 20*np.log10(np.abs(H_hp)), label='RC high-pass')
plt.axvline(fc, linestyle='--', alpha=0.7, label=f'cutoff = {fc:.1f} Hz')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Magnitude [dB]')
plt.title('Magnitude response of first-order RC filters')
plt.grid(True, which='both', alpha=0.3)
plt.legend()

plt.subplot(1, 2, 2)
plt.semilogx(f, np.angle(H_lp, deg=True), label='RC low-pass')
plt.semilogx(f, np.angle(H_hp, deg=True), label='RC high-pass')
plt.axvline(fc, linestyle='--', alpha=0.7, label=f'cutoff = {fc:.1f} Hz')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Phase [deg]')
plt.title('Phase response of first-order RC filters')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```

## 10. Example: a simple RLC band-pass response


<img src="../figs/lec5/RCL_BandPass.svg" width=400/>

A standard second-order band-pass model can be written as

$$
H(s)=\frac{\frac{\omega_0}{Q}s}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}.
$$

This form makes the role of $\omega_0$ and $Q$ explicit:

- $\omega_0$ sets the center frequency,
- $Q$ controls how selective the filter is.

The plot below shows how the resonance becomes sharper as $Q$ increases.

```python
import numpy as np
import matplotlib.pyplot as plt

f0 = 1e3
w0 = 2*np.pi*f0
f = np.logspace(1, 5, 1500)
w = 2*np.pi*f

def H_bp(w, w0, Q):
    s = 1j*w
    return ((w0/Q)*s) / (s**2 + (w0/Q)*s + w0**2)

plt.figure(figsize=(7,4.5))
for Q in [0.5, 1, 2, 5]:
    H = H_bp(w, w0, Q)
    plt.semilogx(f, 20*np.log10(np.abs(H)), label=f'Q={Q}')

plt.axvline(f0, linestyle='--', alpha=0.7, label=f'f0 = {f0:.0f} Hz')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Magnitude [dB]')
plt.title('Band-pass response for different quality factors')
plt.grid(True, which='both', alpha=0.3)
plt.legend()
plt.show()
```

# RLC Band-block

<img src="../figs/lec5/RCL_BandBlock.svg" width=400/>


```python
import numpy as np
import matplotlib.pyplot as plt

# Parametrat e komponentëve
R = 100         # Rezistencë (Ohm)
L = 10e-3       # Induktivitet (H)
C = 1e-6        # Kapacitet (F)

# Frekuencat (omega) në rad/s
omega = np.logspace(2, 6, 1000)
f = omega/2/np.pi

# Impedancat përkatëse
jw = 1j * omega
Z_L = jw * L
Z_C = 1 / (jw * C)
Z_LC = Z_L + Z_C         # Seri LC
Z_R = R
Z_total = Z_R + Z_LC     # Total impedance

# Funksioni i transferimit: Vout merret mbi LC
# H(jw) = Z_LC / (R + Z_LC)
H = Z_LC / Z_total

# Magnituda (dB) dhe faza (gradë)
gain = 20 * np.log10(np.abs(H))
phase = np.angle(H, deg=True)

# Vizatim i grafikëve Bode
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.semilogx(f, gain)
plt.title('Magnitude - Band-block: R - (L + C)')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Gain [dB]')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.semilogx(f, phase)
plt.title('Phase - Band-block: R - (L + C)')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Phase [degree]')
plt.grid(True)

plt.tight_layout()
plt.show()
```

# **Reading: General Analysis of Differential Equations. The Generalized Concept of Impedance $Z$ (Laplace Impedance)** 

#### **Step 1: Apply Kirchhoff’s laws**

In the analysis of electrical circuits with passive elements (**resistor R, capacitor C, inductor L**) and a voltage source $V(t)$ (not necessarily AC), we use Kirchhoff’s laws to write differential equations that describe the behavior of the circuit.

For example, in the RC circuit above, we obtained:
$$ V(t) = RC\frac{dv_C(t)}{dt} + v_C(t) $$

#### **Step 2: Apply the Laplace transform**

The Laplace transform ($\mathcal{L}$) is a powerful method that converts differential equations into algebraic equations:

$$ \mathcal{L}\left\{ \frac{dv_C(t)}{dt} \right\} = s V_C(s)-v_C(0), $$

We explicitly used the properties of the Laplace transform ($s=\sigma +j\omega$):

* $\mathcal{L}\left\{\frac{dv_C}{dt}\right\}= sV_C(s)-v_C(0)$
* $\mathcal{L}\left\{V(t)\right\}=V(s)$

**In the case of an RC circuit with non-zero initial conditions, i.e., when $v_C(0)=V_{C0}$ (meaning the capacitor has voltage $V_{C0}$ at $t=0$)**, the equation becomes:

$$ V(s) = (1+RC,s)V_C(s)-RC,V_{C0} $$

#### **Step 3: Solve the algebraic equation for the capacitor**

We solve explicitly for $V_C(s)$, also defining $\tau=R\, C$:

$$ V_C(s) = \frac{V(s)+\tau\, V_{C0}}{1+\tau \, s} $$

For an explicit solution, consider the case of an AC voltage source:

$$ V(t)=V_0\cos(\omega t)\quad\Longrightarrow\quad V(s)=\frac{sV_0}{s^2+\omega^2} $$

thus, the voltage across the capacitor becomes:

$$ V_C(s)=\frac{\frac{sV_0}{s^2+\omega^2}+\tau\, V_{C0}}{1+\tau\, s} $$

#### **Step 4: Apply the inverse Laplace transform (solution in the time domain)**

The general solution in the time domain is:

$$ V_C(s)=\frac{sV_0}{(1+\tau\, s)(s^2+\omega^2)}+\frac{\tau\, V_{C0}}{1+\tau\,s} $$

where we have separated it into two parts for easier inversion. Taking the inverse transform of the first term:

$$ v_C(t)=\frac{V_0}{1+(\omega\, \tau)^2}\left[\cos(\omega t)+\omega \tau\,\sin(\omega t)\right]-\frac{V_0 e^{-t/\tau}}{1+(\omega\, \tau)^2} $$

This expression contains two components: one corresponding to the **steady-state** solution (purely sinusoidal), and another with exponential terms known as **transient** components. The second term gives:

$$ V_C(s)=\frac{\tau\,V_{C0}}{1+\tau\,s}\quad\Longrightarrow\quad v_C(t)=V_{C0}e^{-t/\tau} $$

This term represents a simple exponentially decaying solution, i.e., a transient term.

#### **Step 5: Full solution**

The complete solution for the capacitor voltage with non-zero initial conditions is:

$$ v_C(t)= \left(V_{C0}- \frac{V_0}{1+(\omega \, \tau)^2} \right) \,e^{-t/\tau} + \frac{V_0}{1+(\omega \,\tau)^2}\left[\cos(\omega t)+\omega \,\tau\,\sin(\omega t)\right] $$

This result clearly shows that the solution contains:

* A transient component due to the initial capacitor voltage
* A transient component due to the application of the source voltage
* The long-term steady-state response after the transient components decay

#### **Special case 1: Solution without source, only initial conditions**

When $V(t)=0$, we simply obtain the natural response:

$$ v_C(t)=V_{C0}\, e^{-t/\tau} $$

#### **Special case 2: Solution with source but no initial conditions**

When $v_C(0)=V_{C0}=0$, and for sufficiently long time $t\gg \tau$, the solution reduces to the steady-state solution:

$$ v_C(t)= \frac{V_0}{1+(\omega \, \tau)^2}\left[\cos(\omega t)+\omega\, \tau \,\sin(\omega t)\right] $$

Defining $A_c=\frac{V_0}{\sqrt{1+(\omega\,\tau)^2}}$ as the amplitude of the capacitor voltage, and $\varphi=\arctan(\omega\,\tau)$ as the phase, we obtain:

$$ v_C(t)= A_c \, \cos(\omega t - \varphi) $$

In other words, if the circuit is driven by a sinusoidal voltage for a sufficiently long time, the effects of initial conditions and transient terms vanish. In an RC circuit, for $t \gg RC=\tau$, only the sinusoidal **steady-state solution** remains. Under these conditions, the Laplace equations reduce to sinusoidal form, i.e., Fourier (AC) analysis where $s = j\omega$. This is the key transition from Laplace analysis to steady-state sinusoidal (Fourier or AC) analysis, which we will use from this point onward.

<br>

<div style="border:5px solid #0a0; background-color:#ca0; color:#003; border-radius: 5px; padding: 10px;">

### **Main assumption for using Fourier analysis:**

1. **Zero initial conditions (i.e., at $t=0$ the capacitor has zero electric field and the inductor has zero magnetic field).**
2. **Sufficiently long time such that the steady state has been reached.**

**Under these conditions, $s=\sigma+j\omega$. Due to the absence of exponential transient terms in the solution as $t\to\infty$, we must have $\sigma\to 0$, therefore $s\to j\omega$.**

## 11. Laboratory connection

This lecture should not remain purely mathematical. In the lab, frequency response is measured directly.

### Basic setup

- function generator for $v_{\text{in}}(t)$,
- oscilloscope channel 1 for input,
- oscilloscope channel 2 for output,
- one RC, RL, or RLC circuit on breadboard.

### What to measure

For a sequence of frequencies:

1. measure the input amplitude,
2. measure the output amplitude,
3. compute
   $$
   |H|=\frac{V_{\text{out}}}{V_{\text{in}}}
   $$
4. estimate the phase shift from the time delay:
   $$
   \phi = -\,2\pi f\,\Delta t.
   $$

### What students should observe

- RC low-pass: strong attenuation at high frequency,
- RC high-pass: strong attenuation at low frequency,
- RLC band-pass: a maximum near resonance,
- phase shift changing systematically with frequency.

This is the experimental meaning of a transfer function.

## 12. Textbook-style remarks and practical cautions

Real filters never behave exactly like ideal textbook blocks.

### Non-ideal effects

- component tolerances shift the cutoff or resonant frequency,
- inductors have winding resistance,
- capacitors may have leakage and parasitic series resistance,
- breadboards and cables add stray capacitance and inductance,
- oscilloscope probes can load the circuit.

Therefore, the measured response is always a combination of:

- the ideal model,
- parasitic effects,
- instrument loading.

That is not a failure of theory. It is part of learning real electronics.

## 13. Summary

- Frequency-domain analysis studies the steady-state response to sinusoids.
- Phasors convert differentiation into multiplication by $j\omega$.
- Impedance generalizes resistance:
  $$
  Z_R=R,\qquad Z_L=j\omega L,\qquad Z_C=\frac{1}{j\omega C}.
  $$
- The transfer function
  $$
  H(j\omega)=\frac{\tilde V_{\text{out}}}{\tilde V_{\text{in}}}
  $$
  encodes both gain and phase shift.
- RC and RL circuits naturally realize low-pass and high-pass filters.
- RLC circuits introduce resonance and quality factor $Q$.
- Integrators and differentiators are limiting regimes of RC filters.
- Bode plots provide a compact visual summary of circuit behavior.
- Laboratory measurement of amplitude ratio and phase shift is the practical verification of this theory.
