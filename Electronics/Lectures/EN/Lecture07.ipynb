# Lecture 7 — Frequency Response & Applications

## 1. Motivation

Instead of solving differential equations, we analyze how circuits respond to different frequencies.


## 2. Sinusoidal Signals

* Sinusoids as fundamental inputs
* Steady-state vs transient response


## 3. Phasor Representation

* Encodes amplitude and phase
* Simplifies sinusoidal analysis



## 4. Impedance

* Generalization of resistance
* $R$, $L$, $C$ in frequency domain



## 5. Transfer Function

$H(\omega) = V_{\rm out}/V_{\rm in}$

This is the central concept for analyzing circuits in frequency domain.



## 6. Filters

### Low-Pass Filter

* Passes low frequencies
* Attenuates high frequencies


### High-Pass Filter

* Passes high frequencies
* Attenuates low frequencies

### Band-Pass Filter

* Passes a range of frequencies



## 7. Integrator & Differentiator

* Special cases of filters
* Practical signal processing applications
